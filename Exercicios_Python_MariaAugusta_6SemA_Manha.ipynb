Calcule o fatorial de um número fornecido pelo usuário.

Para fazer o cálculo, usamos uma variável para armazenar o resultado inicial e um laço de repetição for para percorrer os números do intervalo informado. A cada repetição, o valor atual é multiplicado pelo resultado acumulado, formando o cálculo progressivo do fatorial.

A função int() é utilizada para converter o valor digitado pelo usuário em um número inteiro, permitindo realizar as operações matemáticas. O resultado do fatorial é exibido utilizando a função print().



```python
# Recebemos o número inteiro do usuário
numero = int(input("Digite um número:"))

# Começamos a variável 'fatorial' com 1 (elemento neutro da multiplicação)
fatorial = 1

# O laço vai multiplicar de 1 até o número digitado (range vai até numero+1 para incluir o próprio número)
for valor in range(1, numero +1):
    # Multiplicamos o valor atual pelo que já estava acumulado no fatorial
    fatorial *= valor

# Após o término de todo o laço (fora do 'for', sem recuo!), mostramos o resultado final
print("O fatorial de", numero, "é", fatorial)

```

Exercício 12 – Contagem de Caracteres
Conte os números de ocorrências de cada caractere em uma string fornecida

Usei a estrutura de repetição for para percorrer cada caractere do texto digitado pelo usuário.
Com a estrutura condicional if...else, verificamos a contagem: se o caractere já existir, incrementamos o seu valor em 1 (+=1); caso contrário, ele é inserido  com o valor inicial de 1. Depois, a função print() exibe o resultado da contagem.





```python
# Recebemos o texto digitado pelo usuário
texto = input("Digite uma palavra ou frase: ")

# Criamos um dicionário vazio que vai guardar as letras e suas quantidades
contagem = {}

# O laço 'for' passa por cada letra (caractere) do texto
for caractere in texto:
    # Se a letra já estiver no nosso dicionário, adicionamos +1 na contagem dela
    if caractere in contagem:
        contagem[caractere] += 1
    # Se for a primeira vez que a letra aparece, começamos a contagem dela com 1
    else:
        contagem[caractere] = 1

# Exibimos o dicionário final com o resultado da contagem de todas as letras
print("Contagem de caracteres:", contagem)
```

Exercício 17 - Dicionário de contagem de Palavras
Conte a frequência de cada palavra em uma string fornecida pelo usuário.

Como o texto deve ser fornecido pelo usuário, usamos a função input().
Depois, criamos uma variável (texto).
Para contar as palavras, precisa fatiar a frase, cada vez que tiver um espaço em branco. Então, o método usado é o .split()
o contagem = {} cria um dicionário para armazenar a contagem.
e as condicionais if: Verifica se a palavra já existe no dicionário. Se já existir, somamos 1 ao seu valor atual (acumulando a contagem).
else: Caso a palavra esteja aparecendo pela primeira vez, ela é adicionada ao dicionário com o valor inicial de 1.

```python
# Contar quantas vezes cada palavra aparece em uma frase.
texto = input("Digite uma frase:")

# Usamos o .split() para quebrar a frase nos espaços e criar uma lista de palavras isoladas
palavras = texto.split()

# Dicionário vazio para guardar as palavras e suas contagens
contagem = {}

# Passamos por cada palavra da lista
for palavra in palavras:
    # Se a palavra já foi contada antes, adicionamos mais 1
    if palavra in contagem:
        contagem[palavra] += 1
    # Se for a primeira vez que ela aparece na frase, começamos com 1
    else:
        contagem[palavra] = 1

# AJUSTE DE INDENTAÇÃO: Colocamos este print totalmente fora do laço 'for' (sem recuo)!
# Assim, o Python mostra o resultado final apenas uma vez, em vez de imprimir a cada repetição.
print("Frequência das palavras:", contagem)
```

Exercício 18 – Número Primo.
Crie um programa que verifique se um número fornecido é primo.
Primeiro, pedimos que o usuário digite um número inteiro usando int(input())
.
Para descobrir se ele é primo, começamos assumindo que ele é, criando a variável eh_primo = True.
Depois, verificamos:
Se o número for menor ou igual a 1, ele automaticamente não é primo (eh_primo = False).
Se for maior que 1, usamos um laço for com range(2, numero) para testar se ele é divisível por qualquer outro número. Se o resto de alguma divisão for zero (% i == 0), significa que ele tem outro divisor e não é primo, então usamos o break para parar o teste na hora.
E, usamos a função print() para exibir na tela se o número é primo ou não, baseando-se no resultado final da variável eh_primo.

```python
# Descobrir se um número é Primo (divisível apenas por 1 e por ele mesmo).
numero = int(input("Digite um numero inteiro: "))

# Começamos assumindo que o número é primo (True)
eh_primo = True

# Números menores ou iguais a 1 não são primos por regra matemática
if numero <= 1:
    eh_primo = False
else:
    # Testamos se o número é divisível por qualquer valor entre 2 até o número anterior a ele
    for i in range(2, numero):
        # Se o resto da divisão por algum desses números for zero, significa que ele não é primo
        if numero % i == 0:
            eh_primo = False
            break # Encontramos um divisor, então podemos parar o laço de testes imediatamente (break)

# No final, verificamos o estado da nossa variável de controle 'eh_primo'
if eh_primo:
    print(f"o numero {numero} é primo!")
else:
    print(f"o numero {numero} não é primo!")
```

Implemente um jogo onde o usuário deve adivinhar um número gerado aleatoriamente pelo computador.

Usmos o import random, biblioteca responsável por gerar números aleatórios.
random.randint(1, 100): O computador escolhe um número secreto de 1 a 100.
while: Criamos um loop que continua rodando indefinidamente enquanto o palpite do usuário for diferente do número secreto.
if e elif: Dentro do loop, comparamos o palpite digitado. Se for menor, damos a dica de que o número secreto é maior; se for maior, avisamos que é menor.
print() final: Quando o usuário acerta, o loop while se encerra e o programa exibe a mensagem de parabéns

```python
# Jogo interativo para adivinhar o número que o computador pensou.
# IMPORTANTE: Aqui importamos a biblioteca 'random' do Python,
# que é a caixinha de ferramentas que nos permite gerar e sortear números aleatórios!
import random

# O computador sorteia um número inteiro aleatório entre 1 e 100
numero_secreto = random.randint(1, 100)
print("Eu pensei em um número entre 1 e 100. Tente adivinhar!")

# Começamos a variável de palpite com 0 para que o laço consiga iniciar
palpite = 0

# O jogo continua rodando ENQUANTO (while) o palpite do usuário for diferente do número secreto
while palpite != numero_secreto:
    palpite = int(input("Qual é o seu palpite? "))

    # Damos dicas se o palpite foi muito alto ou muito baixo
    if palpite < numero_secreto:
        print("O número secreto é MAIOR!")
    elif palpite > numero_secreto:
        print("O número secreto é MENOR!")

# Se o laço 'while' terminou, significa que o usuário finalmente acertou!
print(f"Parabéns! Você acertou! O número era {numero_secreto}.")
```

 Exercício 20 – Matriz Transposta.

 Dada uma matriz,(lista de listas), crie sua matriz transposta.

Neste exercício, pegamos uma matriz original (representada por uma lista de listas) e geramos a sua matriz transposta, transformando as linhas em colunas e as colunas em linhas
.
A lógica funciona nos seguintes passos:
Definição da matriz: Criamos uma matriz de exemplo estruturada como uma lista de listas.
Identificação das colunas: Descobrimos a quantidade de colunas que a matriz original possui usando len(matriz).
Loops aninhados (for):
O loop de fora percorre cada coluna da matriz original.
O loop de dentro passa por cada linha, pega o elemento da coluna atual (linha[i]) e o adiciona em uma nova_linha.
transposta.append(): Salva cada nova_linha criada dentro da nossa lista final.
Exibição com print(): Exibe as duas matrizes linha por linha de forma organizada na tela, como uma coluna.

```python
# Transposição de Matriz (transformar as linhas de uma matriz em colunas).
# Definimos nossa matriz original 3x3 (uma lista que guarda outras listas)
matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

# Descobrimos o número de colunas medindo o tamanho da nossa matriz
colunas = len(matriz)

# Criamos uma lista vazia para armazenar a matriz transposta que vamos construir
transposta = []

# O laço 'for' passa pelo índice de cada coluna
for i in range(colunas):
    nova_linha = [] # Criamos uma nova linha vazia para a nossa matriz transposta
    # Passamos por cada linha da matriz original
    for linha in matriz:
        # Pegamos o elemento da coluna 'i' e colocamos na nossa nova linha
        nova_linha.append(linha[i])
    # Adicionamos a linha completada na nossa matriz transposta
    transposta.append(nova_linha)

# AJUSTE DE INDENTAÇÃO: Removemos os prints de dentro dos laços de cálculo!
# Assim as matrizes são exibidas bonitinhas apenas uma vez na tela.
print("Matriz Original:")
for linha in matriz:
    print(linha)

print("\nMatriz Transposta:")
for linha in transposta:
    print(linha)
```

Exercício 21 - Busca Binária

Implemente a busca binária para encontrar um elemento em uma lista ordenada.

Aqui, vamos usar uma lista ordenada e uma elemento de busca . Len, para contar a quantidade de numeros da lista. Definimos inicio, fim e resultado. While é usado para busca . Se o numero procurado é menor que o meio, ele descarta a metade direita, ou vice-versa. E o print() trará o resultado .

```python
# Algoritmo de Busca Binária.
# Em vez de olhar número por número, a busca binária corta a lista organizada ao meio a cada tentativa!
lista = [5, 12, 18, 23, 35, 45, 57, 62, 71, 88]
elemento_procurado = 88

# Definimos os limites da nossa busca (o início e o fim da lista)
inicio = 0
fim = len(lista) - 1
posicao = -1 # Se continuar -1 no final, significa que não encontramos o número

# O laço continua enquanto a nossa área de busca tiver pelo menos um elemento
while inicio <= fim:
    # Achamos o elemento bem no meio da área atual (usamos // para divisão inteira)
    meio = (inicio + fim) // 2

    # Caso 1: Demos sorte e achamos o número logo de primeira!
    if lista[meio] == elemento_procurado:
        posicao = meio
        break # Paramos a busca imediatamente

    # Caso 2: O número do meio é menor do que o procurado.
    # Então, descartamos toda a metade esquerda e focamos na metade direita!
    elif lista[meio] < elemento_procurado:
        inicio = meio + 1

    # Caso 3: O número do meio é maior do que o procurado.
    # Descartamos toda a metade direita e focamos na metade esquerda!
    else:
        fim = meio - 1

# Mostramos o resultado da busca na tela
if posicao != -1:
    print(f"O elemento {elemento_procurado} está na posição {posicao} da lista.")
else:
    print(f"O elemento {elemento_procurado} não foi encontrado na lista.")
```

Crie uma função para criptograr uma mensagem usando a cifra de Cesar.

```python
# Função para criptografar usando a Cifra de César.
# Essa técnica consiste em "empurrar" cada letra do alfabeto algumas posições para a frente.
def criptografar_cesar(mensagem, chave):
    # Definimos o nosso alfabeto padrão e uma string vazia para guardar a mensagem secreta
    alfabeto = "abcdefghijklmnopqrstuvwxyz"
    criptografia = ""

    # Passamos letra por letra da mensagem original
    for letra in mensagem:
        # Deixamos a letra minúscula temporariamente para conseguir achar no nosso alfabeto
        letra_minuscula = letra.lower()

        # Se o caractere for realmente uma letra que está no nosso alfabeto...
        if letra_minuscula in alfabeto:
            # Achamos em qual posição (de 0 a 25) a letra está hoje
            posicao_atual = alfabeto.find(letra_minuscula)

            # Somamos a chave para "andar para a frente". O "% 26" serve para a nossa roda de letras
            # recomeçar do "A" caso passe do "Z" (o alfabeto tem 26 letras)!
            nova_posicao = (posicao_atual + chave) % 26
            nova_letra = alfabeto[nova_posicao]

            # Se a letra original era maiúscula, devolvemos ela maiúscula
            if letra.isupper():
                nova_letra = nova_letra.upper()
                criptografia += nova_letra
            else:
                # Caso contrário, mantemos minúscula
                criptografia += nova_letra
        else:
            # Se for um espaço, vírgula, exclamação ou número, não criptografamos (mantém o original)
            criptografia += letra

    # Devolvemos a mensagem totalmente criptografada
    return criptografia


# --- ÁREA DE TESTE DA FUNÇÃO ---
mensagem_original = "Avante, Dev!"
chave_segredo = 3

resultado = criptografar_cesar(mensagem_original, chave_segredo)
print("--- TESTE DA CIFRA DE CÉSAR ---")
print(f"Original:      {mensagem_original}")
print(f"Criptografada: {resultado}")
```

```python
# Exercício 23: Função para descriptografar (revelar o segredo).
# É o processo exatamente inverso: em vez de somar a chave, nós subtraímos ela para voltar as posições.
def descriptografar_cesar(mensagem_cifrada, chave):
    alfabeto = "abcdefghijklmnopqrstuvwxyz"
    mensagem_original = ""

    # Passamos letra por letra da mensagem secreta
    for letra in mensagem_cifrada:
        letra_minuscula = letra.lower()

        if letra_minuscula in alfabeto:
            posicao_atual = alfabeto.find(letra_minuscula)

            # A ÚNICA MUDANÇA: Subtraímos a chave para fazer o alfabeto "andar para trás"!
            # O "% 26" continua garantindo que o alfabeto funcione como uma roda perfeita.
            nova_posicao = (posicao_atual - chave) % 26
            nova_letra = alfabeto[nova_posicao]

            # Mantemos a formatação maiúscula ou minúscula original
            if letra.isupper():
                mensagem_original += nova_letra.upper()
            else:
                mensagem_original += nova_letra
        else:
            # Mantém espaços, números e símbolos totalmente intactos
            mensagem_original += letra

    # Devolvemos o texto original revelado!
    return mensagem_original


# --- TESTE DO SEU CRIPTOGRAFADOR REVERSO ---
mensagem_secreta = "Dydqwh, Ghy!"  # Mensagem cifrada que geramos no Exercício 22
chave_segredo = 3

resultado_revelado = descriptografar_cesar(mensagem_secreta, chave_segredo)

print("--- DESCRIPTOGRAFIA DA CIFRA DE CÉSAR ---")
print(f"Mensagem Secreta: {mensagem_secreta}")
print(f"Texto Revelado:   {resultado_revelado}")


```

if len(cpf) != 11 or not cpf.isdigit():: Verificamos se o CPF resultante tem exatamente 11 caracteres e se todos eles são dígitos numéricos. Se contiver letras ou tamanho incorreto, o programa retorna False (inválido) imediatamente.
if cpf == cpf[0] * 11:: CPFs compostos por números todos iguais (como 111.111.111-11) são matematicamente "válidos" pelo algoritmo, mas são rejeitados pela Receita Federal por serem falsos. Esse comando multiplica o primeiro caractere por 11 (criando uma sequência repetida) e a compara com o CPF informado. Se forem iguais, o CPF é descartado.
2. Validação do Primeiro Dígito Verificador (10º dígito)
A matemática do CPF funciona multiplicando os dígitos por pesos decrescentes:
soma_1 = 0: Criamos uma variável acumuladora para armazenar a soma das multiplicações.
for i in range(9):: O laço percorre os 9 primeiros dígitos do CPF (índices de 0 a 8).
int(cpf[i]) * (10 - i): Cada dígito é convertido para número inteiro e multiplicado por um peso que começa em 10 e diminui de 1 em 1 a cada rodada (10, 9, 8, ..., 2).
resto_1 = (soma_1 * 10) % 11: A regra oficial manda multiplicar a soma por 10 e extrair o resto da divisão por 11 (operador %).
if resto_1 == 10: resto_1 = 0: Caso o resto seja 10, o regulamento define que o dígito correspondente é 0.
if resto_1 != int(cpf[9]):: O programa compara o resultado calculado com o décimo dígito real do CPF (índice 9). Se não baterem, o código já para e retorna False.
3. Validação do Segundo Dígito Verificador (11º dígito)
A verificação do último número segue uma lógica praticamente idêntica à anterior, porém com duas alterações:
for i in range(10):: O laço agora percorre os 10 primeiros dígitos do CPF, incluindo o primeiro dígito verificador que acabamos de validar.
int(cpf[i]) * (11 - i): Os pesos agora começam em 11 e diminuem progressivamente até 2 (11, 10, 9, ..., 2).
resto_2 = (soma_2 * 10) % 11: Fazemos a mesma conta de resto de divisão por 11.
if resto_2 == 10: resto_2 = 0: Se o resto for 10, ele também se torna 0.
if resto_2 != int(cpf[10]):: Comparamos com o décimo primeiro dígito real do CPF (índice 10). Se forem diferentes, retorna False.
Note que não utilizamos blocos else no código principal. Na programação moderna, estruturar o código usando if seguidos de return rápidos funciona como um funil de segurança. Isso evita o aninhamento excessivo do código (criar "degraus" de indentação para a direita), tornando a leitura e manutenção do programa muito mais simples e limpa.

```python

# Validador completo de CPF.
# Este algoritmo limpa a formatação, bloqueia números falsos repetidos e aplica a fórmula matemática dos dígitos verificadores.
def validar_cpf(cpf):
    # 1. LIMPEZA: Remove os pontos e o traço para trabalhar apenas com os 11 números puros
    cpf = cpf.replace(".", "").replace("-", "")

    # 2. VALIDAÇÃO INICIAL: O CPF precisa ter exatamente 11 caracteres e conter apenas números
    if len(cpf) != 11 or not cpf.isdigit():
        return False # Se não passar nessa regra simples, o CPF é inválido de cara

    # 3. BLOQUEIO DE NÚMEROS REPETIDOS: Evita CPFs falsos conhecidos como "11111111111", "22222222222", etc.
    # (Multiplicamos o primeiro caractere por 11 para comparar se todo o CPF é igual a ele)
    if cpf == cpf[0] * 11:
        return False

    # 4. CÁLCULO DO 1º DÍGITO VERIFICADOR (Posição 9 do CPF)
    soma_1 = 0
    # Multiplicamos os primeiros 9 dígitos pelos pesos de 10 descendo até 2
    for i in range(9):
        soma_1 += int(cpf[i]) * (10 - i)

    # Aplicamos a regra matemática do resto da divisão por 11 multiplicado por 10
    resto_1 = (soma_1 * 10) % 11
    if resto_1 == 10:
        resto_1 = 0

    # Se o dígito calculado for diferente do que está na posição 9 do CPF original, ele é inválido!
    if resto_1 != int(cpf[9]):
        return False

    # 5. CÁLCULO DO 2º DÍGITO VERIFICADOR (Posição 10 do CPF)
    soma_2 = 0
    # Os pesos agora começam em 11 e vão descendo até 2, pegando os 10 primeiros dígitos
    for i in range(10):
        soma_2 += int(cpf[i]) * (11 - i)

    resto_2 = (soma_2 * 10) % 11
    if resto_2 == 10:
        resto_2 = 0

    # Se o segundo dígito calculado for diferente do que está na posição 10 do CPF original, ele é inválido!
    if resto_2 != int(cpf[10]):
        return False

    # Se passou por todas as barreiras matemáticas, o CPF é autêntico e estruturalmente válido!
    return True


# --- ÁREA DE TESTES NO COLAB ---
print("--- TESTANDO O VALIDADOR DE CPF ---")

# Teste 1: CPF Falso com números repetidos
cpf1 = "111.111.111-11"
print(f"O CPF {cpf1} é válido? {validar_cpf(cpf1)}")

# Teste 2: CPF Falso com dígitos inválidos
cpf2 = "123.456.789-09"
print(f"O CPF {cpf2} é válido? {validar_cpf(cpf2)}")

# Teste 3: CPF fictício estruturalmente válido (passa na matemática)
cpf3 = "529.172.930-10"
print(f"O CPF {cpf3} é válido? {validar_cpf(cpf3)}")
```

 Classifique uma lista de números em três categorias : positivos, negativos e zeros.

  Criamos a lista de entrada(lista original) e chamamos a função do lado de fora do def.

```python
# Separador de números positivos, negativos e zeros.
# Criamos uma função que analisa uma lista de números e os separa em três grupos diferentes.
def classificar_numeros(lista):
    # Iniciamos três listas vazias para guardar cada categoria separadamente
    positivos = []
    negativos = []
    zeros = []

    # Passamos por cada número da lista original usando o laço 'for'
    for numero in lista:
        if numero > 0:
            positivos.append(numero)  # Se for maior que zero, adicionamos à lista de positivos
        elif numero < 0:
            negativos.append(numero)  # Se for menor que zero, adicionamos à lista de negativos
        else:
            zeros.append(numero)      # Se não for maior nem menor, só pode ser o próprio zero!

    # Retornamos as três listas prontas e populadas para quem chamou a função
    return positivos, negativos, zeros


# --- ÁREA DE TESTE ---
lista_original = [10, -5, 0, 3, -1, 0, 8, -12]

# Chamamos a função e salvamos o retorno desempacotando nas variáveis pos, neg e zer
pos, neg, zer = classificar_numeros(lista_original)

print("--- RESULTADO DA CLASSIFICAÇÃO ---")
print(f"Lista Original: {lista_original}")
print(f"Positivos:      {pos}")
print(f"Negativos:      {neg}")
print(f"Zeros:          {zer}")
```

Exercicio 27 - Anagramas
Duas palavras são anagramas se tiverem exatamente as mesmas letras reorganizadas em ordens diferentes.
Para padronizar, convertemoos duas palavras para minúsculas, usando o .lower().
Para ordenar em ordem alfabética, usamos a função .sorted() para quebra e  reorganização das plavras e replace() para evitar espaços em branco. Modulamos com def(): Criamos a função verificar_anagrama(palavra1, palavra2) para isolar o algoritmo de teste. Isso deixa o código principal limpo e permite que a verificação de anagramas seja reutilizada facilmente em outros projetos.


Comparação Direta: Usamos o operador de igualdade (==) para comparar as duas listas de letras ordenadas. Se elas forem idênticas, a função retorna True (verdadeiro); se houver qualquer diferença, retorna False (falso).
Usando o comando input().para que o usuario digite e teste qualquer palavra. E o print()para trazer o resultado na tela.

```python
# Verificador de Anagramas.
# Criamos uma função que limpa as palavras e compara se elas usam as mesmas letras.
def verificar_anagrama(palavra1, palavra2):
    # Deixamos tudo em minúsculo (.lower()) e removemos os espaços (.replace)
    # para evitar que letras maiúsculas ou espaços atrapalhem a comparação
    p1 = palavra1.lower().replace(" ", "")
    p2 = palavra2.lower().replace(" ", "")

    # Usamos o sorted() para colocar as letras de cada palavra em ordem alfabética!
    # Se "roma" virar "amor" e "amor" continuar "amor", as duas listas ordenadas serão iguais!
    p1_ordenada = sorted(p1)
    p2_ordenada = sorted(p2)

    # Retorna True se as letras ordenadas forem exatamente iguais, ou False se não forem
    return p1_ordenada == p2_ordenada


# --- ÁREA DE TESTE NO COLAB ---
print("=== VERIFICADOR DE ANAGRAMAS ===")

palavra_usuario1 = input("Digite a primeira palavra: ")
palavra_usuario2 = input("Digite a segunda palavra: ")

# Chamamos a função e mostramos a mensagem correspondente
if verificar_anagrama(palavra_usuario1, palavra_usuario2):
    print(f"'{palavra_usuario1}' e '{palavra_usuario2}' são anagramas!")
else:
    print(f"'{palavra_usuario1}' e '{palavra_usuario2}' não são anagramas.")





```

Exercício 31 - Ordenação por Bolha - Implemente o algoritmo de ordenação por bolha (Bubble Sort).

O algoritmo Bubble Sort tem esse nome porque funciona como "bolhas" de ar na água: os números maiores (mais "leves" ou que devem ir para o final) vão flutuando aos poucos até a última posição da lista.
Usamos um laço for() dentro de outro laço aninhado for() para percorrer varias vezes a lista para comparar e ordenar.
len() para contar quantos numeros existem na lista e o range()cria uma sequencia de passos para o for(), "falando" quantas vezes terá que repetir a comparação dos vizinhos.


```python
# Algoritmo de Ordenação Bubble Sort.
# Este método compara elementos vizinhos e os troca de lugar se estiverem na ordem errada.
# Os maiores números vão "flutuando" para o fim da lista a cada rodada!

# Usamos o 'def' para criar uma função (um atalho inteligente que podemos usar várias vezes!)
def bubble_sort(lista):
    # Descobrimos o tamanho total da lista para saber o limite de elementos
    n = len(lista)

    # O primeiro laço 'for i' controla quantas passadas completas daremos pela lista
    for i in range(n):
        # O segundo laço 'for j' compara cada elemento com o seu vizinho logo à direita (j + 1).
        # Truque inteligente: a cada passada 'i', o maior elemento já "flutuou" para o final.
        # Por isso, usamos 'n - i - 1' para não perder tempo testando quem já está no lugar certo!
        for j in range(0, n - i - 1):

            # Se o número atual (j) for maior do que o seu vizinho da direita (j + 1)...
            if lista[j] > lista[j + 1]:
                # ...eles trocam de lugar!
                # No Python, podemos trocar duas variáveis de lugar de uma vez só:
                # O que estava em j vai para j+1, e o que estava em j+1 vai para j!
                lista[j], lista[j + 1] = lista[j + 1], lista[j]

    # Devolvemos a lista já totalmente organizada para quem chamou a função
    return lista


# --- ÁREA DE TESTE NO COLAB ---
# Criamos uma lista desordenada para testar o algoritmo bolha
numeros = [58, 12, 89, 3, 45, 17, 90, 22]

print("Lista Original Desordenada:", numeros)

# Chamamos a nossa função passando os nossos números desordenados
lista_organizada = bubble_sort(numeros)

# Exibimos a lista totalmente organizada na tela
print("Lista Ordenada com Sucesso: ", lista_organizada)
```

Exercício 32 - Jogo da Velha .
Crie um jogo da velha para dois jogadores

O Tabuleiro (lista de 9 posições): Representamos o tabuleiro como uma lista contendo 9 espaços vazios [' '](índices de 0 a 8). Cada espaço representa uma casa do jogo.
2. As Condições de Vitória: Criamos uma lista de tuplas contendo todas as 8 combinações possíveis de vitória (as 3 linhas, as 3 colunas e as 2 diagonais). A cada jogada, o computador checa se um dos jogadores preencheu alguma dessas combinações.
O Laço while True (Loop Infinito): É o motor do jogo! Ele mantém o jogo rodando sem parar, revezando as jogadas entre "X" e "O", até que aconteça uma das duas condições de parada: "alguém ganhar" ou "dar empate (velha)".
Tratamento de Erros: O código garante que um jogador não jogue em uma casa que já está ocupada ou escolha um número inválido.



```python
# Jogo da Velha completo e interativo.
# Estruturamos o jogo usando funções separadas para desenhar a tela, testar a vitória e rodar a partida!

# 1. FUNÇÃO DE DESENHO: Mostra o tabuleiro formatado na tela para os jogadores
def exibir_tabuleiro(tab):
    # O Python acessa os índices de 0 a 8 para preencher as 9 posições do tabuleiro
    print(f"\n {tab[0]} | {tab[1]} | {tab[2]} ")
    print("---|---|---")
    print(f" {tab[3]} | {tab[4]} | {tab[5]} ")
    print("---|---|---")
    print(f" {tab[6]} | {tab[7]} | {tab[8]} \n")

# 2. FUNÇÃO DE CHECAGEM: Verifica se o jogador atual conseguiu alinhar 3 símbolos
def verificar_vitoria(tab, jogador):
    # Mapeamos as 8 combinações geométricas que dão vitória no jogo (linhas, colunas e diagonais)
    combinacoes_vitoriosas = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8), # Linhas horizontais
        (0, 3, 6), (1, 4, 7), (2, 5, 8), # Colunas verticais
        (0, 4, 8), (2, 4, 6)             # Diagonais cruzadas
    ]

    # Passamos por cada combinação de índices 'c'
    for c in combinacoes_vitoriosas:
        # Se os três espaços daquela combinação tiverem o mesmo símbolo do jogador atual...
        if tab[c[0]] == tab[c[1]] == tab[c[2]] == jogador:
            return True # Temos um vencedor!
    return False # Se passar por todas as combinações e não achar nada, o jogo continua

# 3. FUNÇÃO PRINCIPAL: Controla os turnos, escolhas e as regras do jogo
def jogar():
    # Criamos o tabuleiro inicial com os números de 1 a 9 como strings para guiar os jogadores
    tabuleiro = [str(i) for i in range(1, 10)]
    jogador_atual = "X" # O jogo sempre começa com o jogador 'X'
    jogadas = 0 # Contador para sabermos se deu empate (velha)

    print("=== BEM-VINDA AO JOGO DA VELHA EM PYTHON ===")
    print("Escolha a posição digitando o número correspondente (1 a 9):\n")

    # Laço principal da partida (roda até o jogo acabar por vitória ou empate)
    while True:
        exibir_tabuleiro(tabuleiro)

        # Usamos o bloco try/except para capturar erros caso o usuário digite letras ou números inválidos
        try:
            escolha = int(input(f"Jogador [{jogador_atual}], escolha uma posição livre (1-9): "))
            posicao = escolha - 1 # Ajustamos o número humano (1-9) para o índice do Python (0-8)

            # TRAVA DE SEGURANÇA 1: Impede de escolher uma casa que já tenha sido marcada por "X" ou "O"
            if tabuleiro[posicao] in ["X", "O"]:
                print("⚠️ Essa casa já está ocupada! Escolha outra.")
                continue # Ignora o resto do bloco e repete a pergunta de jogada
        except (ValueError, IndexError):
            # Se der erro de conversão (letras) ou índice fora do range de 0 a 8, avisa o usuário
            print("❌ Entrada inválida! Digite um número de 1 a 9.")
            continue # Reinicia o turno sem quebrar o programa

        # Se passou pelas travas, marca a jogada no tabuleiro com o símbolo do jogador do momento
        tabuleiro[posicao] = jogador_atual
        jogadas += 1 # Registra mais uma jogada realizada

        # Verifica se essa última jogada trouxe a vitória
        if verificar_vitoria(tabuleiro, jogador_atual):
            exibir_tabuleiro(tabuleiro)
            print(f"🎉 PARABÉNS! O Jogador [{jogador_atual}] venceu o jogo!")
            break # Encerra o laço e finaliza o programa

        # Se chegamos a 9 jogadas e ninguém venceu, deu empate (velha)
        if jogadas == 9:
            exibir_tabuleiro(tabuleiro)
            print("👵 Deu VELHA! O jogo terminou empatado.")
            break # Encerra a partida

        # ALTERNÂNCIA DE JOGADORES: Se era "X" vira "O", se era "O" vira "X"
        # Usamos um if simplificado em uma única linha (operador ternário)
        jogador_atual = "O" if jogador_atual == "X" else "X"

# Inicia o jogo chamando a nossa função principal!
jogar()
```

Exercício 33 - Verificação de Anagramas - Verifique se duas palavras fornecidas são anagramas uma da outra.

Duas palavras são anagramas se usarmos exatamente as mesmas letras para escrever ambas, apenas mudando a ordem delas.
Usaremos aqui as seguintes funções : lower(), para transformar todas as letras em minúsculas. replace(), para remover qualquer espaço em branco.len(), para comparar o tamanho das palavras, quantidades de caracteres. sorted(), para organizar as letras das palavras em ordem alfabética e, se forem idênticas, confirmar que é um anagrama.  



```python
# Verificação de Anagramas.
# Criamos uma função inteligente que limpa os textos, compara o tamanho das palavras
# e ordena as letras para verificar se elas são idênticas.

def verificar_anagrama(palavra1, palavra2):
    # Removemos os espaços (.replace) e deixamos todas as letras minúsculas (.lower())
    p1 = palavra1.lower().replace(" ", "")
    p2 = palavra2.lower().replace(" ", "")

    # Teste de tamanho: se as duas palavras limpas têm comprimentos diferentes,
    # o Python já retorna False imediatamente e economiza processamento.
    if len(p1) != len(p2):
        return False

    # Se passarem no teste de tamanho, ordenamos as letras de cada uma em ordem alfabética (sorted)
    # e comparamos se as listas resultantes são idênticas.
    return sorted(p1) == sorted(p2)


# --- ÁREA DE TESTE NO COLAB ---
palavra_a = input("Digite a primeira palavra: ")
palavra_b = input("Digite a segunda palavra: ")

# Chamamos a nossa função e mostramos a resposta correta na tela
if verificar_anagrama(palavra_a, palavra_b):
    print(f"'{palavra_a}' e '{palavra_b}' são anagramas!")
else:
    print(f"'{palavra_a}' e '{palavra_b}' não são anagramas.")
```

```python
# Contador de Vogais.
# Criamos uma função que analisa um texto e conta quantas vogais (a, e, i, o, u) existem nele.

def contar_vogais(texto):
    # Começamos o acumulador em zero para iniciar a contagem
    contador = 0

    # Definimos quais são as letras que estamos procurando (nossas vogais)
    vogais = "aeiou"

    # PADRONIZAÇÃO: Transformamos todo o texto em letras minúsculas.
    # Assim, se o usuário digitar "PYTHON", o programa lê "python" e consegue contar as vogais sem problemas!
    texto_padronizado = texto.lower()

    # Passamos letra por letra do texto padronizado usando o laço 'for'
    for letra in texto_padronizado:
        # Se a letra atual estiver contida na nossa string de vogais...
        if letra in vogais:
            # ...adicionamos mais 1 ao nosso acumulador (este bloco precisa de recuo extra!)
            contador += 1

    # Retornamos o total acumulado de vogais após analisar todo o texto
    return contador


# --- ÁREA DE TESTE NO COLAB ---
# Coletamos o texto do usuário
frase_usuario = input("Digite uma frase ou palavra: ")

# Chamamos a função passando o texto coletado e guardamos o resultado
total_vogais = contar_vogais(frase_usuario)

# Exibimos o resultado final formatado na tela
print(f"A frase/palavra '{frase_usuario}' possui {total_vogais} vogais.")
```

Exercício 35 - Números Fibonacci Até N- Gere a sequência de Fibonacci até um número n fornecido pelo usuário

A sequência de Fibonacci é uma ordem matemática onde cada número novo é o resultado da soma dos dois números anteriores. Para iniciar a sequência, precisamos de dois valores iniciais obrigatórios: o 0 e o 1.

Neste exercício, o objetivo é gerar essa sequência parando antes de ultrapassar um limite N digitado pelo usuário. Para que o computador faça isso, estruturamos o programa assim:
1. Lista Inicial: Criamos uma lista contendo os dois primeiros números: [0, 1].
2. Laço while (Enquanto): Usamos esse laço de repetição para calcular novos números continuamente, pois não sabemos de antemão quantos números precisaremos gerar até chegar ao limite.
3. Soma dos Últimos ([-1] e [-2]): Usamos os índices negativos do Python para pegar de forma simples o último ([-1]) e o penúltimo ([-2]) número da lista e somá-los.
4. Regra de Parada: A cada rodada, o programa testa se a nova soma passou do limite N definido pelo usuário. Se passou, o laço é interrompido imediatamente com o comando break, e a lista final é exibida.





```python

# Esta lógica é incrível porque ela não calcula uma quantidade fixa de termos,
# mas sim continua gerando números até atingir o limite estipulado pelo usuário.

def gerar_fibonacci_ate_n(n):
    # TRAVAS DE SEGURANÇA (Tratamento de casos especiais):
    # Se o limite for um número negativo, retornamos uma lista vazia
    if n < 0:
        return []
    # Se o limite for zero, a sequência só contém o primeiro termo [0]
    if n == 0:
        return [0]
    # Se o limite for um, a sequência para no segundo termo [0, 1]
    if n == 1:
        return [0, 1]

    # Começamos a nossa lista de números com os dois primeiros termos clássicos: 0 e 1
    sequencia = [0, 1]

    # Criamos um laço de repetição infinito (while True) que controlaremos manualmente lá dentro
    while True:
        # TRUQUE SENSACIONAL DO PYTHON:
        # O índice [-1] acessa o ÚLTIMO item da lista e o [-2] acessa o PENÚLTIMO!
        # Somando os dois, temos exatamente o próximo número da sequência de Fibonacci.
        proximo_numero = sequencia[-1] + sequencia[-2]

        # CRITÉRIO DE PARADA: Se o próximo número que calculamos for maior do que o limite "n",
        # nós paramos o laço infinito na hora com o comando 'break'!
        if proximo_numero > n:
            break

        # Se o número passou no teste de limite, adicionamos ele no final da nossa lista (.append)
        sequencia.append(proximo_numero)

    # Devolvemos a lista de números gerada até o limite
    return sequencia


# --- ÁREA DE TESTE NO COLAB ---
# Coletamos o número limite inteiro digitado pelo usuário
limite_n = int(input("Digite um número limite para a sequência de Fibonacci: "))

# Chamamos a função e passamos o limite definido
resultado = gerar_fibonacci_ate_n(limite_n)

# Exibimos o resultado final formatado na tela
print(f"Sequência de Fibonacci até {limite_n}: {resultado}")
```

Exercício 37 - Funções Recursivas

Implemente uma função recursiva para calcular o fatorial de um número.

Função recursiva é simplesmente uma função que chama a si mesma.
O fatorial de um número é o próprio  número multiplicado pelo fatorial de seu antecessor(n-1). E o fatorial de 0 ou 1 é igual a 1.
def fatorial(n):: Define a receita do fatorial, onde n é o número de entrada que queremos calcular
.
if n == 0 or n == 1:: É o nosso Caso Base (a regra de segurança ou ponto de parada)
. Ele faz uma pergunta crucial ao computador: "O número atual é 0 ou 1?". Se a resposta for Sim (Verdadeiro), o programa para de criar novos ajudantes e devolve diretamente o valor 1
.
Por que ele é vital? Sem esse if para dizer "chega!", a fábrica continuaria chamando ajudantes para sempre (fatorial de -1, -2, -3...) e o computador travaria em um loop infinito.
else:: É o nosso Plano de Ação para todos os números maiores
. Se a pergunta do if for Não (Falso) — ou seja, se o número ainda for grande, como 4 —, o computador ignora a regra de parada e cai obrigatoriamente aqui
.
return n * fatorial(n - 1): É a Chamada Recursiva
. É aqui que a mágica acontece. O ajudante atual não faz a conta inteira; ele apenas guarda o seu próprio número (n *) e chama um novo ajudante do andar de baixo para resolver uma tarefa um pouquinho menor: o fatorial do número anterior fatorial(n - 1)
.
return: Devolve o resultado final de cada multiplicação para o ajudante do andar de cima, até que a resposta chegue na dona da fábrica e seja exibida para o cliente





```python
# Cálculo de Fatorial usando Recursão.
# A recursão acontece quando uma função chama a ela mesma para resolver partes menores do problema!

def fatorial(n):
    # CRITÉRIO DE PARADA (Caso Base): Toda função recursiva precisa de uma saída de segurança.
    # Como o fatorial de 0 ou de 1 é sempre 1, se chegarmos aqui, a função para de se chamar
    # e começa a devolver os resultados de volta.
    if n == 0 or n == 1:
        return 1

    # CASO RECURSIVO: Se o número for maior que 1, a função chama a si mesma
    # passando o número anterior (n - 1) e multiplica pelo número atual (n).
    # Exemplo: fatorial(3) vira: 3 * fatorial(2) -> que vira: 2 * fatorial(1)...
    else:
        return n * fatorial(n-1)


# --- ÁREA DE TESTE NO COLAB ---
# Coletamos o número inteiro desejado pelo usuário
numero = int(input("Digite um número para calcular seu fatorial: "))

# Chamamos a nossa função recursiva passando o número digitado
resultado = fatorial(numero)

# Exibimos o resultado final calculado na tela
print(f"O fatorial de {numero} é: {resultado}")

```

Exercício 42 - Números Primos até N

Gere todos os números primos até um número n fornecido  

Aqui iremos gerar numeros primos até um limite n.
os primos, enquanto não sabemos, ficará em lista vazia []
usamos o laço for() para testar qual numero testar por vez. um segundo laço for() tenta dividir o numero da vez por vários valores menores que ele, chamado de divisor. Se o  operador %(resto de divisão)for zero, descobrimos que a divisão é exata, ou seja, o número não é primo. Nesse caso, a plaquinha(fictícia) muda para False e o break interrompe os testes desse número.
Se o teste terminar e a plaquinha(fictícia), variável booleana:que seria como um interruptor, True ou False , continuar como True, significa que o número é realmente primo, e usamos o .append() para guardá-lo na nossa lista. No final, o print() exibe de forma simples, usando vírgulas, o limite definido e a lista final de primos."

```python
# Um número primo é aquele que só pode ser dividido por 1 e por ele mesmo.
# Usamos dois laços de repetição para testar as divisões de cada número!

# 1. Definimos até qual número queremos fazer a nossa busca
limite = 20

# 2. Criamos uma lista vazia para armazenar os números primos que encontrarmos
primos = []

# 3. LAÇO EXTERNO: Passamos por cada número a partir de 2 até o nosso limite.
# (Começamos no 2 porque os números 0 e 1 não são considerados primos pela matemática)
for numero in range(2, limite + 1):
    # Criamos uma variável de controle (booleana) chamada 'primo'.
    # Começamos acreditando que o número atual É primo até que se prove o contrário!
    primo = True

    # LAÇO INTERNO: Testamos se o nosso número atual é divisível por qualquer valor
    # que vá desde 2 até o número imediatamente anterior a ele (numero - 1).
    for divisor in range(2, numero):
        # Usamos o operador de resto da divisão (%) para testar a divisão exata.
        # Se o resto for igual a 0, significa que o número tem outro divisor além dele mesmo!
        if numero % divisor == 0:
            # Se encontramos um divisor, mudamos nossa variável de controle para False...
            primo = False
            # ...e interrompemos o laço interno com 'break', pois não precisamos mais continuar testando!
            break

    # Se a nossa variável continuou como 'True' após passar por todos os testes de divisão,
    # significa que o número é realmente primo!
    if primo:
        # Adicionamos o número primo encontrado na nossa lista de resultados
        primos.append(numero)

# 4. Exibimos a lista completa de números primos encontrados até o limite definido
print("Números primos até", limite, ":", primos)
```

Exercício 44 - Sequência de Fibonacci

Gere a sequência de Fibonacci até um determinado  números de termos.

Neste exercício, o objetivo é gerar a sequência de Fibonacci até um limite de termos definido no código (que escolhemos como 10).
Começamos criando uma lista vazia [] para receber a sequência e definimos as duas variáveis de partida: anterior = 0 e atual = 1.
Usamos o laço for com o caractere _ (que apenas repete o código a quantidade de vezes que pedimos, sem precisar de uma variável de contagem). A cada rodada do laço, usamos o .append() para salvar o valor anterior na nossa lista.
Logo em seguida, fazemos a fórmula matemática: somamos o anterior com o atual para descobrir o proximo número. Depois, atualizamos as variáveis para que elas dêem um "passo à frente" para a próxima rodada: o anterior recebe o valor do atual e o atual recebe o valor do proximo.
No final, o print() trará o resultado final com a lista completa exibido uma única vez no console

```python
# Sequência de Fibonacci com número fixo de termos.
# Este programa calcula exatamente a quantidade de termos da sequência que você definir!

# 1. Definimos quantos números da sequência nós queremos gerar
total_termos = 5

# 2. Criamos uma lista vazia que vai guardar a sequência conforme ela for gerada
sequencia_fibonacci = []

# 3. Iniciamos as variáveis com as "sementes" (os primeiros valores da sequência)
anterior = 0
atual = 1

# 4. Criamos um laço de repetição 'for' que vai rodar exatamente a quantidade de vezes definida.
# Usamos o caractere "_" porque não precisamos usar o número da contagem dentro do laço.
for _ in range(total_termos):
    # Adicionamos o número 'anterior' na nossa lista usando o método .append()
    sequencia_fibonacci.append(anterior)

    # Fazemos a soma matemática para descobrir o próximo termo da sequência
    proximo = anterior + atual

    # Atualizamos as variáveis para a próxima rodada (as variáveis dão um "passo à frente"!)
    anterior = atual
    atual = proximo

# 5. Exibimos o resultado final contendo a nossa lista de termos gerada
print("Os primeiros", total_termos, "termos de Fibonacci são:", sequencia_fibonacci)
```

Exercício 46 - Calcular Média e Desvio Padrão

Calcule a média e o desvio padrão de uma lista de números

Neste exercício, calculamos a média e o desvio padrão de uma lista de valores diretamente no código. Primeiro, descobrimos a quantidade de itens na lista usando o comando len(). Em seguida, usamos um laço for para somar todos os números e dividimos essa soma pela quantidade, encontrando a media.
Para calcular o desvio padrão, usamos um segundo laço for que calcula a diferença de cada número em relação à média e eleva essa diferença ao quadrado (usando ** 2) para eliminar valores negativos, acumulando tudo na variável soma_diferenca_quadrado.
Dividimos esse total acumulado pela quantidade de números para encontrar a variancia. Por fim, tiramos a raiz quadrada da variância elevando-a a 0.5 para obter o desvio_padrao final. Os resultados são exibidos na tela de forma limpa usando o print(). Ainda não entendi direito esse exercício......

```python
# Exercício: Cálculo de Média e Desvio Padrão.
# Analisamos um conjunto de dados para descobrir a média e o quanto os valores variam em relação a ela.

# 1. Definimos a nossa lista com os valores numéricos que vamos analisar
valores = [5, 10, 13, 15, 18, 20, 25, 30]

# 2. Descobrimos a quantidade total de elementos na lista
quantidade = len(valores)

# --- ETAPA 1: CÁLCULO DA MÉDIA ---
# Criamos um acumulador para somar todos os números
soma_total = 0

# Usamos um laço de repetição 'for' para passar por cada número e somá-los
for numero in valores:
    soma_total = soma_total + numero

# Dividimos a soma total pela quantidade de itens para achar a média aritmética
media = soma_total / quantidade


# --- ETAPA 2: CÁLCULO DA VARIÂNCIA ---
# Criamos outro acumulador para somar o quadrado das diferenças
soma_diferenca_quadrado = 0

# Passamos novamente pela lista de valores com um laço 'for'
for numero in valores:
    # Descobrimos a distância (diferença) entre o número atual e a média que calculamos
    diferenca = numero - media

    # Elevamos essa diferença ao quadrado (usando o operador '** 2') para eliminar valores negativos
    diferenca_quadrado = diferenca ** 2

    # Acumulamos o resultado na nossa variável de soma
    soma_diferenca_quadrado = soma_diferenca_quadrado + diferenca_quadrado

# A variância é a média dessas diferenças elevadas ao quadrado
variancia = (soma_diferenca_quadrado / quantidade)


# --- ETAPA 3: CÁLCULO DO DESVIO PADRÃO ---
# O desvio padrão é simplesmente a raiz quadrada da variância.
# TRUQUE MATEMÁTICO: No Python, elevar um número a '0.5' (pow 0.5) é o mesmo que calcular a sua raiz quadrada!
desvio_padrao = variancia ** 0.5


# --- EXIBIÇÃO DOS RESULTADOS ---
print(f"A média dos valores é: {media}")
print(f"O desvio padrão dos valores é: {desvio_padrao:.4f}")
```

Exercício 47 - Contar numeros em lista

Conte quantas vezes cada número aparece em uma lista

definimos uma lista de números e criamos um dicionário vazio chamado contagem = {}. O dicionário é ideal aqui porque ele nos permite associar cada número (chave) à sua respectiva quantidade de aparições (valor).
Usamos um laço for para percorrer a lista original. Para cada número, usamos a estrutura condicional if: se o número ainda não estiver no dicionário (if numero not in contagem), nós o criamos lá dentro com o valor inicial de 1. Caso contrário, o else é ativado e nós apenas somamos mais 1 ao valor que já estava guardado.
No final, usamos um segundo laço for junto com o método .items() para extrair simultaneamente o número e a sua quantidade do dicionário, exibindo os resultados formatados na tela através do print(), que mostrará o resultado.



```python
# Contador de Frequência de Números.
# Usamos um dicionário para registrar quantas vezes cada número aparece dentro de uma lista.

# 1. Definimos a nossa lista de números com vários valores repetidos
lista = [1, 2, 4, 5, 2, 1, 6, 9, 7, 8, 0, 5, 5, 8, 9, 3, 3, 8, 3]

# 2. Criamos um dicionário vazio.
# Ele será o nosso "painel de controle". A CHAVE será o número e o VALOR será a quantidade de repetições.
contagem = {}

# 3. Usamos um laço de repetição 'for' para analisar cada número da lista, um por um
for numero in lista:
    # Se o número ainda NÃO foi registrado como uma chave no nosso dicionário...
    if numero not in contagem:
        # ...nós o adicionamos lá e dizemos que ele apareceu 1 vez (iniciamos o contador dele)
        contagem[numero] = 1
    # Se o número já estiver registrado lá dentro...
    else:
        # ...nós acessamos o valor atual dele e somamos mais 1 à contagem
        contagem[numero] += 1

# 4. Usamos outro laço 'for' para exibir os resultados de forma organizada.
# O método '.items()' é fantástico porque nos devolve duas coisas de uma vez só:
# a chave (numero) e o valor associado a ela (quantidade)!
for numero, quantidade in contagem.items():
    print(f"O número {numero} aparece {quantidade} vezes.")
```

Exercício 48 - Ordenação por Inserção

Implemente o algorítmo de ordenção por inserção(Insert Sort)

Aqui nesse exercício, comparamos ao jogo de cartas. Com as cartas nas mãos[], e verificando que há desordem entre elas,
usamos o for() para percorrê-las a partir da segunda carta, guardando a carta que queremos encaixar, junto com o range(), que gera uma sequencia de posições começando a partir de 1(segunda carta), até o final da lista. O range inicia em 1, pois a primeira carta(índice 0) jé é considerada organizada por estar sozinha na mão. Guardamos a carta da vez na variável carta_atual.
Com um laço while(), comparamos essa carta_atual com as anteriores à esquerda(controladas pela variável posição_comparação). Enquanto as cartas da esquerda forem maiores que a nossa carta_atual, elas são "empurradas" uma posição para a direita. .Assim , o espaço é aberto, a carta_atual é inserida na posição correta e o print() exibe as cartas organizadas em ordem crescente.




```python
# Este algoritmo simula o jeito que organizamos cartas na mão, encaixando uma por uma no lugar correto.

# 1. Definimos a nossa lista de "cartas" totalmente desordenada
cartas = [64, 34, 25, 12, 22, 11, 90]

# 2. LAÇO EXTERNO: Começamos a percorrer a lista a partir da SEGUNDA carta (índice 1).
# A primeira carta (índice 0) nós assumimos que já está na posição correta por enquanto.
for indice_atual in range(1, len(cartas)):
    # Guardamos a carta atual que vamos tentar "encaixar" no lado esquerdo organizado
    carta_atual = cartas[indice_atual]

    # Definimos o índice da carta imediatamente à esquerda para começar a comparação
    posicao_comparacao = indice_atual - 1

    # LAÇO INTERNO (while): Ele vai rodar enquanto:
    # - Não chegarmos no início da lista (posicao_comparacao >= 0)
    # - E a carta da esquerda for maior do que a nossa carta atual
    while posicao_comparacao >= 0 and cartas[posicao_comparacao] > carta_atual:
        # "Empurramos" a carta maior uma posição para a direita para abrir espaço
        cartas[posicao_comparacao + 1] = cartas[posicao_comparacao]

        # Damos um passo para trás (para a esquerda) para testar a próxima carta
        posicao_comparacao = posicao_comparacao - 1

    # Assim que a condição do 'while' falhar (achamos o espaço correto),
    # colocamos a nossa carta atual na vaga que foi aberta!
    cartas[posicao_comparacao + 1] = carta_atual

# 3. Exibimos o resultado final com as cartas perfeitamente ordenadas de forma crescente
print("Cartas ordenadas:", cartas)
```

Usamos o laço for() para percorrer cada dígito dentro de dígitos e , dentro do laço, usamos o int() para converter temporariamente o caractere do dígito de volta em um número inteiro e adicionamos esse valor na nossa variável soma. E o print(), trará o resultado da soma de todos os dígitos do número.  


```python
# Uma solução super inteligente que trata o número como texto para conseguir
# passar por cada algarismo individualmente usando o laço 'for', sem precisar de contas complexas!

# 1. Coletamos o número como texto (string).
# Não usamos o 'int(input())' aqui de propósito para podermos tratar o número como uma palavra!
digitos = input("Digite um numero inteiro: ")

# 2. Criamos uma variável acumuladora para guardar a soma dos algarismos
soma = 0

# 3. O laço 'for' passa por cada caractere (cada dígito) do texto digitado
for digito in digitos:
    # Como o 'digito' atual ainda é um texto, nós o convertemos temporariamente
    # para número inteiro usando int() e o somamos ao nosso acumulador
    soma = soma + int(digito)

# 4. Exibimos o resultado final calculado
print("A soma dos seus digitos é:", soma)


```

Exercício 50 - Cálculo de Frequência de Letras

Calcule a frequência de cada letra em uma string fornecida pelo usuário.

Nesse exercício, o usuario digite uma frase ou texto, input() e guardamos na variável  texto. Criamos um dicionário vazio{} para guardar a contagem de cada letra : frequencia{}. Usamos for() é para percorrer o texto.  
Dentro do laço, usamos um if para garantir que os espaços em branco não sejam contados. Em seguida, verificamos se a letra já existe no nosso dicionário: se ela não estiver lá, ela é adicionada com o valor inicial de 1; se ela já existir, o else apenas adiciona mais 1 à sua contagem.
No final,o print()  exibe na tela o dicionário com o resultado final da frequência de cada letra.

```python
# Usamos um dicionário para contar quantas vezes cada letra aparece em uma frase,
# ignorando os espaços em branco para limpar o nosso resultado.

# 1. Coletamos o texto ou a frase do usuário
texto = input("Digite um texto ou frase: ")

# 2. Criamos um dicionário vazio que vai registrar a nossa contagem
frequencia = {}

# 3. Usamos o laço 'for' para passar por cada caractere do texto, um por um
for letra in texto:
    # SACADA DE MESTRE: Verificamos se o caractere atual NÃO é um espaço em branco.
    # Isso evita que o espaço seja contado como se fosse uma letra!
    if letra != " ":
        # Se a letra ainda não foi registrada no dicionário, iniciamos a contagem dela com 1
        if letra not in frequencia:
            frequencia[letra] = 1
        # Se ela já estiver lá, apenas somamos mais 1 ao acumulador dela
        else:
            frequencia[letra] += 1

# 4. Exibimos o dicionário resultante com a frequência de cada letra encontrada
print(f"A frequência de cada letra no texto é: {frequencia}")
```
