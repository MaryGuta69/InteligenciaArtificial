Neste exercício, pegamos uma matriz original (representada por uma lista de listas) e geramos a sua matriz transposta, transformando as linhas em colunas e as colunas em linhas
.
A lógica funciona nos seguintes passos:
Definição da matriz: Criamos uma matriz de exemplo estruturada como uma lista de listas.
Identificação das colunas: Descobrimos a quantidade de colunas que a matriz original possui usando len(matriz).
Loops aninhados (for):
O loop de fora percorre cada coluna da matriz original.
O loop de dentro passa por cada linha, pega o elemento da coluna atual (linha[i]) e o adiciona em uma nova_linha.
transposta.append(): Salva cada nova_linha criada dentro da nossa lista final.
Exibição com print(): Exibe as duas matrizes linha por linha de forma organizada na tela, como uma coluna.

```python
matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
colunas = len(matriz)
transposta = []
for i in range(colunas):
    nova_linha = []
    for linha in matriz:
        nova_linha.append(linha[i])
    transposta.append(nova_linha)

    print("Matriz Original:")
for linha in matriz:
    print(linha)

print("\nMatriz Transposta:")
for linha in transposta:
    print(linha)
```

Exercício 21 - Busca Binária

Implemente a busca binária para encontrar um elemento em uma lista ordenada.

Aqui, vamos usar uma lista ordenada e uma elemento de busca . Len, para contar a quantidade de numeros da lista. Definimos inicio, fim e resultado. While é usado para busca . Se o numero procurado é menor que o meio, ele descarta a metade direita, ou vice-versa. E o print() trará o resultado .

```python
lista = [5, 12, 18, 23, 35, 45, 57, 62, 71, 88]
elemento_procurado = 62


inicio = 0
fim = len(lista) - 1
posicao = -1

while inicio <= fim:
    meio = (inicio + fim) // 2
    if lista[meio] == elemento_procurado:
        posicao = meio
        break
    elif lista[meio] < elemento_procurado:
        inicio = meio + 1
    else:
        fim = meio - 1

if posicao != -1:
    print(f"O elemento {elemento_procurado} está na posição {posicao} da lista.")
else:
    print(f"O elemento {elemento_procurado} não foi encontrado na lista.")
```

Exercicio 23 - Descriptografar Cifra de César

Crie uma função para descriptografar uma mensagem usando a cifra de Cesar.

Este exercício faz o caminho inverso do anterior
. Criamos a função descriptografar_cesar que recebe o texto secreto e a chave.
A lógica é exatamente a mesma da criptografia, com a única diferença na matemática: subtraímos a chave (posicao_atual - chave) para "voltar as casas" no alfabeto e revelar o texto original.

```python
def descriptografar_cesar(mensagem_cifrada, chave):
    alfabeto = "abcdefghijklmnopqrstuvwxyz"
    mensagem_original = ""

    for letra in mensagem_cifrada:
        letra_minuscula = letra.lower()

        if letra_minuscula in alfabeto:
            posicao_atual = alfabeto.find(letra_minuscula)

            # A ÚNICA MUDANÇA: Subtraímos a chave em vez de somar!
            nova_posicao = (posicao_atual - chave) % 26
            nova_letra = alfabeto[nova_posicao]

            # Mantém a formatação maiúscula
            if letra.isupper():
                mensagem_original += nova_letra.upper()
            else:
                mensagem_original += nova_letra
        else:
            # Mantém espaços, números e símbolos intactos
            mensagem_original += letra

    return mensagem_original


# --- TESTE DO SEU CRIPTOGRAFADOR REVERSO ---
mensagem_secreta = "Dydqwh, Ghy!"  # Esta é a mensagem que geramos no Exercício 22!
chave_segredo = 3

resultado_revelado = descriptografar_cesar(mensagem_secreta, chave_segredo)

print("--- DESCRIPTOGRAFIA DA CIFRA DE CÉSAR ---")
print(f"Mensagem Secreta: {mensagem_secreta}")
print(f"Texto Revelado:   {resultado_revelado}")

```

if resto_2 != int(cpf[10]):: Comparamos com o décimo primeiro dígito real do CPF (índice 10). Se forem diferentes, retorna False.

```python
def validar_cpf(cpf):
    # 1. LIMPEZA: Remove pontos e traços
    cpf = cpf.replace(".", "").replace("-", "")

    # 2. VALIDAÇÃO INICIAL: Tamanho de 11 dígitos e apenas números
    if len(cpf) != 11 or not cpf.isdigit():
        return False

    # 3. BLOQUEIO DE NÚMEROS REPETIDOS: Evita "11111111111", "22222222222", etc.
    if cpf == cpf[0] * 11:
        return False

    # 4. CÁLCULO DO 1º DÍGITO VERIFICADOR (Posição 9 do CPF)
    soma_1 = 0
    for i in range(9):
        soma_1 += int(cpf[i]) * (10 - i)

    resto_1 = (soma_1 * 10) % 11
    if resto_1 == 10:
        resto_1 = 0

    if resto_1 != int(cpf[9]):
        return False  # Primeiro dígito não bateu, CPF inválido

    # 5. CÁLCULO DO 2º DÍGITO VERIFICADOR (Posição 10 do CPF)
    soma_2 = 0
    for i in range(10):
        # Os pesos agora começam em 11 e vão descendo até 2
        soma_2 += int(cpf[i]) * (11 - i)

    resto_2 = (soma_2 * 10) % 11
    if resto_2 == 10:
        resto_2 = 0

    if resto_2 != int(cpf[10]):
        return False  # Segundo dígito não bateu, CPF inválido

    # Se passou por todas as barreiras sem retornar False, o CPF é válido!
    return True


# --- ÁREA DE TESTES NO COLAB ---
print("--- TESTANDO O VALIDADOR DE CPF ---")

# Teste 1: CPF Falso com números repetidos
cpf1 = "111.111.111-11"
print(f"O CPF {cpf1} é válido? {validar_cpf(cpf1)}")

# Teste 2: CPF Falso com dígitos inválidos
cpf2 = "123.456.789-09"
print(f"O CPF {cpf2} é válido? {validar_cpf(cpf2)}")

# Teste 3: CPF fictício estruturalmente válido (passa na matemática)
cpf3 = "529.172.930-10"
print(f"O CPF {cpf3} é válido? {validar_cpf(cpf3)}")

```

Exercício 31 - Ordenação por Bolha - Implemente o algoritmo de ordenação por bolha (Bubble Sort).

O algoritmo Bubble Sort tem esse nome porque funciona como "bolhas" de ar na água: os números maiores (mais "leves" ou que devem ir para o final) vão flutuando aos poucos até a última posição da lista.
Usamos um laço for() dentro de outro laço aninhado for() para percorrer varias vezes a lista para comparar e ordenar.
len() para contar quantos numeros existem na lista e o range()cria uma sequencia de passos para o for(), "falando" quantas vezes terá que repetir a comparação dos vizinhos.


```python
#def serve para criar um atalho inteligente, deixando o código mais limpo, organizado e fácil de corrigir!
def bubble_sort(lista):
    n = len(lista)

    # Loop para controlar quantas vezes vamos passar pela lista
    for i in range(n):
        # Loop para comparar os elementos vizinhos
        # (reduzimos o final em '- i' porque os maiores já foram "flutuando" para o fim)
        for j in range(0, n - i - 1):

            # Se o elemento atual for maior que o vizinho da direita...
            if lista[j] > lista[j + 1]:
                # ...eles trocam de lugar na memória!
                lista[j], lista[j + 1] = lista[j + 1], lista[j]

    return lista

# Lista de teste totalmente desordenada
numeros = [58, 12, 89, 3, 45, 17, 90, 22]


print("Lista Original Desordenada:", numeros)

# Chamamos a função para organizar a lista
lista_organizada = bubble_sort(numeros)

print("Lista Ordenada com Sucesso: ", lista_organizada)
```

Tratamento de Erros: O código garante que um jogador não jogue em uma casa que já está ocupada ou escolha um número inválido.



```python
def exibir_tabuleiro(tab):
    # Desenha o tabuleiro na tela mostrando cada casa (de 0 a 8) individualmente
    print(f"\n {tab[0]} | {tab[1]} | {tab[2]} ")
    print("---|---|---")
    print(f" {tab[3]} | {tab[4]} | {tab[5]} ")
    print("---|---|---")
    print(f" {tab[6]} | {tab[7]} | {tab[8]} \n")

def verificar_vitoria(tab, jogador):
    # Lista com as 8 combinações vencedoras (linhas, colunas e diagonais)
    combinacoes_vitoriosas = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8), # Linhas
        (0, 3, 6), (1, 4, 7), (2, 5, 8), # Colunas
        (0, 4, 8), (2, 4, 6)             # Diagonais
    ]

    # CORREÇÃO AQUI: Acessamos cada índice da combinação 'c' individualmente
    for c in combinacoes_vitoriosas:
        if tab[c[0]] == tab[c[1]] == tab[c[2]] == jogador:
            return True
    return False

def jogar():
    # Criamos o tabuleiro inicial com os números de 1 a 9 para guiar o usuário
    tabuleiro = [str(i) for i in range(1, 10)]
    jogador_atual = "X"
    jogadas = 0

    print("=== BEM-VINDA AO JOGO DA VELHA EM PYTHON ===")
    print("Escolha a posição digitando o número correspondente (1 a 9):\n")

    while True:
        exibir_tabuleiro(tabuleiro)

        try:
            escolha = int(input(f"Jogador [{jogador_atual}], escolha uma posição livre (1-9): "))
            posicao = escolha - 1 # Ajusta para os índices do Python (0 a 8)

            # Valida se a casa escolhida já está marcada por X ou O
            if tabuleiro[posicao] in ["X", "O"]:
                print("⚠️ Essa casa já está ocupada! Escolha outra.")
                continue
        except (ValueError, IndexError):
            print("❌ Entrada inválida! Digite um número de 1 a 9.")
            continue

        # Marca a jogada no tabuleiro
        tabuleiro[posicao] = jogador_atual
        jogadas += 1

        # Verifica se o jogador atual ganhou
        if verificar_vitoria(tabuleiro, jogador_atual):
            exibir_tabuleiro(tabuleiro)
            print(f"🎉 PARABÉNS! O Jogador [{jogador_atual}] venceu o jogo!")
            break

        # Se deu 9 jogadas e ninguém ganhou, deu empate (velha)
        if jogadas == 9:
            exibir_tabuleiro(tabuleiro)
            print("👵 Deu VELHA! O jogo terminou empatado.")
            break

        # Alterna o jogador: se era X vira O, se era O vira X
        jogador_atual = "O" if jogador_atual == "X" else "X"

# Inicia o jogo!
jogar()
```

Exercício 35 - Números Fibonacci Até N- Gere a sequência de Fibonacci até um número n fornecido pelo usuário

A sequência de Fibonacci é uma ordem matemática onde cada número novo é o resultado da soma dos dois números anteriores. Para iniciar a sequência, precisamos de dois valores iniciais obrigatórios: o 0 e o 1.

Neste exercício, o objetivo é gerar essa sequência parando antes de ultrapassar um limite N digitado pelo usuário. Para que o computador faça isso, estruturamos o programa assim:
1. Lista Inicial: Criamos uma lista contendo os dois primeiros números: [0, 1].
2. Laço while (Enquanto): Usamos esse laço de repetição para calcular novos números continuamente, pois não sabemos de antemão quantos números precisaremos gerar até chegar ao limite.
3. Soma dos Últimos ([-1] e [-2]): Usamos os índices negativos do Python para pegar de forma simples o último ([-1]) e o penúltimo ([-2]) número da lista e somá-los.
4. Regra de Parada: A cada rodada, o programa testa se a nova soma passou do limite N definido pelo usuário. Se passou, o laço é interrompido imediatamente com o comando break, e a lista final é exibida.





```python
def gerar_fibonacci_ate_n(n):
  if n < 0:
    return []
  if n == 0:
    return [0]
  if n == 1:
    return [0, 1]

  sequencia = [0, 1]
  while True:
      proximo_numero = sequencia[-1] + sequencia[-2]
      if proximo_numero > n:
        break
      sequencia.append(proximo_numero)
  return sequencia

limite_n = int(input("Digite um número limite para a sequência de Fibonacci: "))
resultado = gerar_fibonacci_ate_n(limite_n)

print(f"Sequência de Fibonacci até {limite_n}: {resultado}")
```

 for i in range(tamanho). A letra i serve para contar as rodadas (rodada 0, rodada 1, rodada 2...).

```python
import random
import string

def gerar_senha_aleatoria(tamanho):
  caracteres = string.ascii_letters + string.digits
  senha = ''.join(random.choice(caracteres) for _ in range(tamanho))
  return senha

tamanho_senha = int(input("Digite o tamanho da senha desejada: "))
senha_gerada = gerar_senha_aleatoria(tamanho_senha)
print(f"Sua senha gerada é: {senha_gerada}")
```

Exercício 37 - Funções Recursivas

Implemente uma função recursiva para calcular o fatorial de um número.

Função recursiva é simplesmente uma função que chama a si mesma.
O fatorial de um número é o próprio  número multiplicado pelo fatorial de seu antecessor(n-1). E o fatorial de 0 ou 1 é igual a 1.
def fatorial(n):: Define a receita do fatorial, onde n é o número de entrada que queremos calcular
.
if n == 0 or n == 1:: É o nosso Caso Base (a regra de segurança ou ponto de parada)
. Ele faz uma pergunta crucial ao computador: "O número atual é 0 ou 1?". Se a resposta for Sim (Verdadeiro), o programa para de criar novos ajudantes e devolve diretamente o valor 1
.
Por que ele é vital? Sem esse if para dizer "chega!", a fábrica continuaria chamando ajudantes para sempre (fatorial de -1, -2, -3...) e o computador travaria em um loop infinito.
else:: É o nosso Plano de Ação para todos os números maiores
. Se a pergunta do if for Não (Falso) — ou seja, se o número ainda for grande, como 4 —, o computador ignora a regra de parada e cai obrigatoriamente aqui
.
return n * fatorial(n - 1): É a Chamada Recursiva
. É aqui que a mágica acontece. O ajudante atual não faz a conta inteira; ele apenas guarda o seu próprio número (n *) e chama um novo ajudante do andar de baixo para resolver uma tarefa um pouquinho menor: o fatorial do número anterior fatorial(n - 1)
.
return: Devolve o resultado final de cada multiplicação para o ajudante do andar de cima, até que a resposta chegue na dona da fábrica e seja exibida para o cliente





```python
def fatorial(n):
  if n == 0 or n == 1:
    return 1

  else:
    return n * fatorial(n-1)

numero = int(input("Digite um número para calcular seu fatorial: "))
resultado = fatorial(numero)
print(f"O fatorial de {numero} é: {resultado}")

```

Exercício 40 - Lista de Compras - Crie um programa para gerenciar uma lista de compras, permitindo adicionar, remover e listar itens.

Usamos aqui 4 ferramentas fundamentais do Python:

.append() para adicionar itens no final da lista
.
.remove() para buscar e retirar um item específico pelo nome
.
in para perguntar ao Python se um item está lá dentro antes de tentar apagá-lo (evitando erros)
.
enumerate() para listar tudo numerado e bem bonito na tela



```python
lista_compras =[]
while True:
  print("\n=== Lista de Compras ===")
  print("1. Adicionar item")
  print("2. Remover item")
  print("3. Listar itens")
  print("4. Sair")
  opcao = input("Escolha uma opção (1/2/3/4): ")

  if opcao == "1":
    novo_item = input("Digite o nome do item a ser adicionado:")
    if novo_item !="":
      lista_compras.append(novo_item)
      print(f"{novo_item} foi adicionado à lista de compras.")
    else:
      print("Você não digitou nada!")
  elif opcao == "2":
    if len(lista_compras) == 0:
      print("A lista de compras está vazia.")
    else:
      item_remover = input("Digite o nome do item a ser removido: ")
      if item_remover in lista_compras:
        lista_compras.remove(item_remover)
        print(f"{item_remover} foi removido da lista de compras.")
      else:
        print(f"{item_remover} não está na lista de compras.")
  elif opcao == "3":
    if len(lista_compras) == 0:
      print("A lista de compras está vazia.")
    else:
      print("\nItens na lista de compras:")
      for indice, item in enumerate(lista_compras, 1):
        print(f"{indice}. {item}")
  elif opcao == "4":
    print("Saindo do programa. Até a próxima!")
    break
  else:
    print("Opção inválida. Por favor, escolha 1, 2, 3 ou 4.")
```

Exercício 42 - Números Primos até N

Gere todos os números primos até um número n fornecido  

Aqui iremos gerar numeros primos até um limite n.
os primos, enquanto não sabemos, ficará em lista vazia []
usamos o laço for() para testar qual numero testar por vez. um segundo laço for() tenta dividir o numero da vez por vários valores menores que ele, chamado de divisor. Se o  operador %(resto de divisão)for zero, descobrimos que a divisão é exata, ou seja, o número não é primo. Nesse caso, a plaquinha(fictícia) muda para False e o break interrompe os testes desse número.
Se o teste terminar e a plaquinha(fictícia), variável booleana:que seria como um interruptor, True ou False , continuar como True, significa que o número é realmente primo, e usamos o .append() para guardá-lo na nossa lista. No final, o print() exibe de forma simples, usando vírgulas, o limite definido e a lista final de primos."

```python
limite = 20
primos = []

for numero in range(2, limite + 1):
  primo = True
  for divisor in range(2, numero):
    if numero % divisor == 0:
      primo = False
      break
  if primo:
    primos.append(numero)

print("Números primos até", limite, ":", primos)
```

Exercício 44 - Sequência de Fibonacci

Gere a sequência de Fibonacci até um determinado  números de termos.

Neste exercício, o objetivo é gerar a sequência de Fibonacci até um limite de termos definido no código (que escolhemos como 10).
Começamos criando uma lista vazia [] para receber a sequência e definimos as duas variáveis de partida: anterior = 0 e atual = 1.
Usamos o laço for com o caractere _ (que apenas repete o código a quantidade de vezes que pedimos, sem precisar de uma variável de contagem). A cada rodada do laço, usamos o .append() para salvar o valor anterior na nossa lista.
Logo em seguida, fazemos a fórmula matemática: somamos o anterior com o atual para descobrir o proximo número. Depois, atualizamos as variáveis para que elas dêem um "passo à frente" para a próxima rodada: o anterior recebe o valor do atual e o atual recebe o valor do proximo.
No final, o print() trará o resultado final com a lista completa exibido uma única vez no console

```python
total_termos = 5
sequencia_fibonacci = []
anterior = 0
atual = 1

for _ in range(total_termos):
  sequencia_fibonacci.append(anterior)
  proximo = anterior + atual
  anterior = atual
  atual = proximo

print("Os primeiros", total_termos, "termos de Fibonacci são:" , sequencia_fibonacci)
```

Para calcular o desvio padrão, usamos um segundo laço for que calcula a diferença de cada número em relação à média e eleva essa diferença ao quadrado (usando ** 2) para eliminar valores negativos, acumulando tudo na variável soma_diferenca_quadrado.
Dividimos esse total acumulado pela quantidade de números para encontrar a variancia. Por fim, tiramos a raiz quadrada da variância elevando-a a 0.5 para obter o desvio_padrao final. Os resultados são exibidos na tela de forma limpa usando o print(). Ainda não entendi direito esse exercício......

```python
valores = [5, 10, 13, 15, 18, 20, 25, 30]
quantidade = len(valores)
soma_total = 0
for numero in valores:
  soma_total = soma_total + numero
media = soma_total / quantidade

soma_diferenca_quadrado = 0
for numero in valores:
  diferenca = numero - media
  diferenca_quadrado = diferenca ** 2
  soma_diferenca_quadrado = soma_diferenca_quadrado + diferenca_quadrado

variancia = (soma_diferenca_quadrado / quantidade)
desvio_padrao = variancia ** 0.5

print(f"A média dos valores é: {media}")
print(f"O desvio padrão dos valores é: {desvio_padrao}")
```

Exercício 48 - Ordenação por Inserção

Implemente o algorítmo de ordenção por inserção(Insert Sort)

Aqui nesse exercício, comparamos ao jogo de cartas. Com as cartas nas mãos[], e verificando que há desordem entre elas,
usamos o for() para percorrê-las a partir da segunda carta, guardando a carta que queremos encaixar, junto com o range(), que gera uma sequencia de posições começando a partir de 1(segunda carta), até o final da lista. O range inicia em 1, pois a primeira carta(índice 0) jé é considerada organizada por estar sozinha na mão. Guardamos a carta da vez na variável carta_atual.
Com um laço while(), comparamos essa carta_atual com as anteriores à esquerda(controladas pela variável posição_comparação). Enquanto as cartas da esquerda forem maiores que a nossa carta_atual, elas são "empurradas" uma posição para a direita. .Assim , o espaço é aberto, a carta_atual é inserida na posição correta e o print() exibe as cartas organizadas em ordem crescente.




```python
cartas = [64, 34, 25, 12, 22, 11, 90]
for indice_atual in range(1, len(cartas)):
  cartas_atual = cartas[indice_atual]
  posicao_comparacao = indice_atual - 1
  while posicao_comparacao >= 0 and cartas[posicao_comparacao] > cartas_atual:
    cartas[posicao_comparacao + 1] = cartas[posicao_comparacao]
    posicao_comparacao = posicao_comparacao - 1
  cartas[posicao_comparacao + 1] = cartas_atual

print("Cartas ordenadas:", cartas)
```
