Exercício 1 - Array Unidimensional de 50 elementos
Crie um array unidimensional de 50 elementos contendo números inteiros aleatórios entre 10 e 100.
Unidimensional (1D - Uma Linha): É uma fila única de elementos. Não tem "altura" ou "andares" — é apenas uma sequência reta de dados colocados lado a lado.

Exercício 2 - Gere uma matriz 4X4 com números inteiros de 1 a 16 e extraia a última linha.

Matriz (Bidimensional - 2D): Em vez de uma "fila única" como no Exercício 1, agora vamos criar uma tabela organizada em linhas e colunas. Para dizer ao NumPy que queremos essa estrutura de tabela, usamos o parâmetro size=(4, 4) (o que significa: crie 4 linhas e 4 colunas, totalizando 16 números sorteados).
E para extrair a ultima linha , sem ficar contando manualmente, o Python usa o índice [-1] : um número negativo dentro dos colchetes.


```python
import numpy as np

# Geramos uma sequência de números inteiros de 1 a 16
# - np.arange(1, 17) cria os inteiros de 1 a 16 (o teto 17 é exclusivo)
# - .reshape(4, 4) reorganiza essa sequência em uma matriz de 4 linhas e 4 colunas
matriz = np.arange(1, 17).reshape(4, 4)

print("Matriz 4x4 original:")
print(matriz)
print("-" * 50)  # Nossa linha divisória visual

# Extraímos a última linha da matriz usando fatiamento (slicing)
# - O índice '-1' seleciona a última linha de trás para frente.
# - O ':' (dois pontos) garante que queremos todas as colunas daquela linha.
ultima_linha = matriz[-1, :]

print("Última linha extraída:")
print(ultima_linha)
```

Exercício 4 - Gere um array de 100 números inteiros aleatórios e encontre a mediana e a moda dos valores
Mediana: É o valor que fica exatamente no meio quando organizamos todos os 100 números em ordem crescente
Moda: É o número que mais se repete dentro do nosso grupo de dados. Como o NumPy puro não possui uma função direta para calcular a moda, nós podemos importar uma biblioteca padrão do próprio Python chamada statistics (que já vem instalada e pronta para uso) e usar a função mode().
O Sorteador (np.random.randint)
O que ele faz: Seria como um saco preto com bolinhas numeradas e que o computador coloque a mão lá dentro e sorteie 100 números para nós.
O tamanho (size=100): É o limite de 100 bolinhas para serem retiradas de dentro do saco.
Os limites das bolinhas (1, 101): As bolinhas no saco vão de 1 a 100. Colocamos o limite final como 101 porque o NumPy sempre para um número antes do que digitamos





```python
import numpy as np
import statistics  # Importamos uma biblioteca auxiliar simples para calcular a moda

# Sorteamos as 100 bolinhas do saco (de 1 a 100)
dados = np.random.randint(1, 101, size=100)

print("Bolinhas sorteadas de dentro do saco:")
print(dados)
print("-" * 50)

# Organizamos a fila e pegamos quem está bem no meio
mediana = np.median(dados)

# Descobrimos qual número foi o campeão de repetições
moda = statistics.mode(dados)

print("Resultados da nossa análise:")
print(f"Mediana (o meio da fila): {mediana}")
print(f"Moda (o que mais se repetiu): {moda}")
```

Exercício 5 - Crie um array de 20 elementos contendo números entre 1 e 100 e substitua todos os valores múltiplos de 3 por -3.
Aqui vamos usar o sorteador , nosso saco de bolinhas e tirar 20 bolinhas. (size=20). Como queremos numeros entre 1 e 100, colocamos o limite 101, pois o NumPy sempre para um número antes.
Os multiplos de 3  serão "achados" através de: numeros[numeros % 3 == 0] = -3. Ele acha e transforma em -3 na hora.




```python
import numpy as np

# Sorteamos 20 bolinhas de 1 a 100 de dentro do nosso saco
# O limite 101 garante que o número 100 possa ser sorteado
numeros = np.random.randint(1, 101, size=20)

print("Array original com as 20 bolinhas:")
print(numeros)
print("-" * 50)

# Passamos um pente fino para encontrar os múltiplos de 3 (resto da divisão % por 3 igual a zero)
# Todos os números encontrados são substituídos por -3 de uma vez só
numeros[numeros % 3 == 0] = -3

print("Array modificado (os múltiplos de 3 viraram -3):")
print(numeros)
```

Exercício 9 - Crie um array com 10 números aleatórios e normalize seus valores entre 0 e 1.

O que é Normalizar? Imagine que o computador sorteia 10 números aleatórios de tamanhos bem diferentes. Normalizar é o ato de "espremer" ou "esticar" esses números para que todos eles caibam perfeitamente dentro de uma régua que vai de 0 a 1.
O menor número sorteado do grupo é obrigado a virar exatamente 0.0.
O maior número sorteado do grupo é obrigado a virar exatamente 1.0.
Todos os outros números do meio vão se ajeitar de forma proporcional entre o 0 e o 1.

Como fazemos essa conta? É uma fórmula matemática simples: nós pegamos cada número, subtraímos o menor valor do grupo e dividimos o resultado pela diferença entre o maior e o menor valor do grupo.
A fórmula na prática: (dados - minimo) / (maximo - minimo)

```python
import numpy as np

# Sorteamos 10 números inteiros aleatórios de 1 a 100 de dentro do nosso saco de bolinhas
dados = np.random.randint(1, 101, size=10)

print("Array original com os 10 números sorteados:")
print(dados)
print("-" * 50)

# Descobrimos quem é o menor e quem é o maior número do nosso grupo
minimo = np.min(dados)
maximo = np.max(dados)

# Aplicamos a fórmula de normalização para espremer todos os números entre 0 e 1
dados_normalizados = (dados - minimo) / (maximo - minimo)

print("Array normalizado (tudo espremido proporcionalmente entre 0 e 1):")
print(dados_normalizados)
```

Exercício 10 - Crie uma matriz identidade 6x6 e substitua a diagonal principal por valores crescentes de 1 a 6.

A Matriz Identidade (np.eye(6)): É como uma tabela quadrada (de 6 linhas e 6 colunas) cheia de números 0, mas que tem uma "estrada" de números 1 que corta ela bem na diagonal, de cima à esquerda até embaixo à direita.
A escada de números (np.arange(1, 7)): Criamos uma sequência de números que vai crescer de 1 a 6. Colocamos o limite final como 7 porque o NumPy é teimoso e sempre para um número antes do que a gente digita.
Substituindo a diagonal (np.fill_diagonal): Em vez de a gente ir trocando cada número na mão (o que daria um trabalho danado!), o NumPy tem um comando mágico que localiza essa "estrada" diagonal da tabela e troca todos os valores antigos pelos nossos novos números de 1 a 6 de uma vez só!


```python
import numpy as np

# Criamos uma matriz identidade de tamanho 6x6
# Ela nasce cheia de zeros e com a nossa estrada de 1s na diagonal
matriz = np.eye(6)

print("Matriz Identidade original (com a estrada de 1s):")
print(matriz)
print("-" * 50)

# Criamos a nossa escadinha de números de 1 a 6
valores_novos = np.arange(1, 7)

# O NumPy localiza a diagonal e troca todos os valores de uma vez só
np.fill_diagonal(matriz, valores_novos)

print("Matriz modificada (com a estrada de 1 a 6):")
print(matriz)
```

Exercício 11 - Gere uma array de 100 números inteiros e encontre o valor que mais se repete no array.

Aqui vamos fazer o sorteio de 100 bolinhas (np.random.randint)
Contando as bolinhas (np.unique): Passamos um "pente fino" para separar os números que apareceram e contar, com um cronômetro, quantas vezes cada um foi sorteado.
A bolinha campeã (np.argmax): Encontramos a posição de quem teve a maior contagem de todas. No seu sorteio real, a bolinha 96 foi a grande campeã, sendo sorteada exatamente 4 vezes!



```python
import numpy as np

# Sorteamos 100 bolinhas de 1 a 100 de dentro do nosso saco
bolinhas = np.random.randint(1, 101, size=100)

print("Array com as 100 bolinhas sorteadas:")
print(bolinhas)
print("-" * 50)

# O np.unique nos devolve os valores sorteados e quantas vezes cada um apareceu
valores, contagens = np.unique(bolinhas, return_counts=True)

# Descobrimos a posição (o índice) onde está a maior contagem de todas
indice_campeao = np.argmax(contagens)

# Usamos essa posição para encontrar qual foi o número e quantas vezes ele apareceu
valor_mais_repetido = valores[indice_campeao]
quantidade_repeticoes = contagens[indice_campeao]

print(f"O número que mais se repetiu no saco foi o: {valor_mais_repetido}")
print(f"Ele foi sorteado exatamente {quantidade_repeticoes} vezes!")
```

Exercício 13 - Crie uma array contendo os 10 primeiros números pares e calcule sua soma cumulativa .

Vamos criar aqui (com np.arange) o que seria tipo uma "escada", listando os 10 primeiros números pares positivos (começando no 2 e indo até o 20). Com o np.cumsum(), fazemos a soma cumulativa, que funciona como uma bola de neve ladeira abaixo: começa com o peso de 2, no próximo degrau se junta com o 4 e passa a valer 6 (2+4), e assim sucessivamente até chegar ao final. E o print() trará a soma acumulada passo a passo!

```python
import numpy as np

# Criamos a nossa escada com os 10 primeiros números pares
# Começamos no 2, colocamos o limite em 21 e pulamos de 2 em 2
numeros_pares = np.arange(2, 21, 2)

print("Os 10 primeiros números pares da nossa escada:")
print(numeros_pares)
print("-" * 50)

# Usamos o cumsum para criar o efeito bola de neve somando tudo pelo caminho
soma_acumulada = np.cumsum(numeros_pares)

print("A soma acumulada acumulando o peso passo a passo (bola de neve):")
print(soma_acumulada)

```

Exercício 14 - Calcule a diferença entre dois arrays aleatórios de mesmo comprimento e encontre a média das diferenças.

Usamos aqui o np.random: sacudir o saco e sortear dois grupos independentes(Grupo A e Grupo B), com o mesmo número de elementos cada um.
Diferença entres eles (subtração): Aqui, quando escrevemos grupo_a - grupo_b, o NumPy pega o primeiro número do grupo A e subtrai pelo primeiro número do grupo B , depois faz assim com os segundos, terceiros, etc...tudo no mesmo momento !
Como são números aleatórios, usamos o comando np.mean(diferenças) para calcular a média de todos. E print() para trazer o resultado .

```python
import numpy as np

# Sorteamos dois grupos de números decimais aleatórios com 10 elementos cada
grupo_a = np.random.rand(10)
grupo_b = np.random.rand(10)

print("Grupo A sorteado:")
print(grupo_a)
print("-" * 50)

print("Grupo B sorteado:")
print(grupo_b)
print("-" * 50)

# Calculamos a diferença elemento por elemento de uma só vez
diferencas = grupo_a - grupo_b

print("Diferenças obtidas elemento por elemento:")
print(diferencas)
print("-" * 50)

# Encontramos a média de todas essas diferenças
media_diferencas = np.mean(diferencas)

print(f"A média das diferenças entre os dois grupos é: {media_diferencas}")
```

Exercício 15 -  Gere um array contendo 20 números e substitua todos os valores maiores que 50 por 50.

Sorteio de numeros np.random.randint() , exemplo de 1 a 100.
Pense que 50 é o limite máximo permitido. Qualquer bolinha sorteada que tiver um número maior que 50 (como 78, 95 ou 51) será obrigada a "bater no teto" e virar exatamente 50. Quem for menor ou igual a 50, permanece com seu número. Existe uma função espetacular chamada np.where(condição, valor_se_for_verdade, valor_se_for_mentira), para checagem da informação. E o print() para trazer o resultado .



```python
import numpy as np
# Sorteamos 20 números inteiros de 1 a 100 de dentro do nosso saco
dados = np.random.randint(1, 101, size=20)

print("Array original com as 20 bolinhas sorteadas:")
print(dados)
print("-" * 50)

# Usamos o np.where como um guarda de trânsito inteligente:
# Se o número for maior que 50, ele vira 50. Se não, continua sendo ele mesmo!
dados_limitados = np.where(dados > 50, 50, dados)

print("Array modificado (qualquer número acima de 50 virou 50):")
print(dados_limitados)

```

Exercício 17 - Gere uma matriz 4x4 de números inteiros aleatórios e multiplique cada valor por sua posiçãp matriz.
Começamos sorteando uma matriz 4x4 com números inteiros aleatórios de 1 a 100 de dentro do nosso saco usando np.random.randint(). Depois, para representar as posições (ou endereços) de cada casinha, criamos uma lista de 0 a 15 usando o np.arange(16) (já que uma tabela 4x4 tem 16 posições) e reorganizamos ela com o reshape(4, 4) para transformar em linhas e colunas . Por fim, o NumPy faz a mágica de multiplicar cada número da matriz sorteada pela sua respectiva posição na tabela, tudo de uma única vez e no mesmo momento! E os print() trarão os resultados na tela

```python
import numpy as np

# Sorteamos a matriz 4x4 com números aleatórios de 1 a 100 de dentro do nosso saco
sorteio = np.random.randint(1, 101, size=(4, 4))

print("Matriz 4x4 de números inteiros aleatórios:")
print(sorteio)
print("-" * 50)

# Criamos a nossa tabela de endereços de 0 a 15 (já que uma tabela 4x4 tem 16 posições)
# E usamos o reshape para organizar os dados, transformando a fila única em linhas e colunas 4x4
posicoes = np.arange(16).reshape(4, 4)

print("Tabela com o endereço de cada posição (0 a 15):")
print(posicoes)
print("-" * 50)

# O NumPy faz a mágica de multiplicar cada número sorteado pela sua posição correspondente no mesmo momento
resultado = sorteio * posicoes

print("Matriz final (cada número multiplicado pela sua posição):")
print(resultado)
```

Exercíco 18 - Crie um array de 10 elementos contendo números entre 1 e 100 e substitua todos os valores múltiplos de 5 por 0.
Sorteio de números (np.random.randint): Usamos o np.random.randint() para sacudir o saco e sortear 10 números inteiros entre 1 e 100.
Identificando os múltiplos de 5: Para saber se um número é múltiplo de 5, usamos o resto da divisão em Python (o símbolo %). Se o resto de um número dividido por 5 for igual a zero, significa que ele é múltiplo de 5 (como 15, 30, 45, etc.).
O "Guarda de Trânsito" (np.where): Usamos a função espetacular np.where(condição, valor_se_for_verdade, valor_se_for_mentira) para fazer a checagem da informação no mesmo momento: onde o número for múltiplo de 5 ele é obrigado a virar 0, e onde não for, ele permanece com o seu valor original! E o print() trará o resultado na tela



```python
import numpy as np

# Sorteamos 10  números inteiros de 1 a 100 dentro do nosso saco
elementos = np.random.randint(1, 101, size=10)

print("Array original sorteado:")
print(elementos)
print("-" * 50)

# Usamos o np.where como  um guarda de trânsito inteligente:
# Se o resto da divisão por 5 for zero, o número vira 0. Se não, continua ele mesmo!
elementos_modificados = np.where(elementos % 5 == 0, 0, elementos)

print("Array modificado (qualquer múltiplo de 5 virou 0):")
print(elementos_modificados)

```

```python
import numpy as np
# 1. Sorteamos dois arrays de tamanho 20 com números de 1 a 100 (como você começou!)
array_1 = np.random.randint(1, 101, size=20)
array_2 = np.random.randint(1, 101, size=20)

print("Array 1 sorteado:")
print(array_1)
print("-" * 50)

print("Array 2 sorteado:")
print(array_2)
print("-" * 50)

# 2. Calculamos o produto escalar (o cruzamento de forças dos dois arrays)
produto_escalar = np.dot(array_1, array_2)

# 3. Calculamos o tamanho (norma) de cada um dos arrays na memória do computador
tamanho_1 = np.linalg.norm(array_1)
tamanho_2 = np.linalg.norm(array_2)

# 4. Dividimos o cruzamento pelo tamanho deles para achar a sintonia de direção (cosseno)
cosseno = produto_escalar / (tamanho_1 * tamanho_2)

# 5. Descobrimos o ângulo de inclinação em radianos (sua linha de código brilhante!)
angulo_radianos = np.arccos(cosseno)

# 6. Convertemos o resultado para graus para ficar bem fácil de ler
angulo_graus = np.degrees(angulo_radianos)

print(f"Cosseno (sintonia de direção): {cosseno:.4f}")
print(f"Ângulo em radianos: {angulo_radianos:.4f}")
print(f"O ângulo real entre os dois arrays é de: {angulo_graus:.2f} graus")
```

Primeiro, usei o NumPy para gerar cinco números complexos de forma aleatória (com a semente 42, para garantir que os valores fiquem fixos) misturando partes reais e imaginárias com o j. Exercício 16 - Resolva a equação diferencial de segunda ordem frac{d^2y}{dx^2} + y = 0 , com condições iniciais  y(0) = 1  e  y'(0) = 0

Na física , essa equação representa o movimento de um pêndulo simples ou de uma mola oscilando sem atrito.
No Python, o submódulo scipy.integrate possui uma ferramenta incrível chamada solve_ivp (solve initial value problem) especializada em resolver equações diferenciais.
Para resolver uma equação de segunda ordem , usamos um truque : dividimos a equação original em um sistema de duas equações de segunda ordem
No Python, o submódulo scipy.integrate possui uma ferramenta incrível chamada solve_ivp (solve initial value problem) especializada em resolver equações diferenciais


```python
import numpy as np
from scipy.integrate import solve_ivp

# Definimos o sistema de duas equações de primeira ordem
# u representa o y, e u[4] representa as derivadas dy
def sistema(x,u):
  y, dy = u
  return [dy, -y] # Retorna as derivadas correspondentes: [y', y'']

# Definimos o intervalo de busca para x (de 0 a 10)
x_span=[0, 10]

# Definimos as condições iniciais: y(0) = 1e y'(0) = 0
condições_iniciais = [1.0 , 0.0]

# Criamos os pontos de x onde queremos calcular os resultados
x_pontos = np.linspace(0, 10, 100)

#Resolvemos a equação usando o solve_ivp
solucao = solve_ivp(sistema, x_span, condições_iniciais, t_eval=x_pontos)

# Exibimos os primeiros resultados calculaos para y
solucao.y

# Mostra apenas as 5 primeiras respostas de y
solucao.y[0][:5]
```

Spline cúbica é como se fosse uma régua maleável, que conecta os pontos, deixando-as sem quinas.
Então usaremos o submódulo scipy.interpolate e a ferramenta CubicSpline. Defini os pontos de partida de X de 1 a 5 e calculei seus respectivos valores de Y elevando-os ao quadrado
. Em seguida, utilizei a classe CubicSpline do módulo scipy.interpolate para gerar uma curva suave e contínua conectando esses pontos. Para testar o modelo, escolhi pontos que estavam exatamente no meio dos valores originais ([1.5, 2.5, 3.5, 4.5]) e o programa calculou os novos valores de Y de forma rápida e com precisão perfeita.

```python
import numpy as np
from scipy.interpolate import CubicSpline

# Definimos os pontos conhecidos de x e calculamos y = x²
x_pontos = np.array([1, 2, 3, 4, 5]) # Corrigido para ter múltiplos pontos
y_pontos = x_pontos**2  # y será [1, 4, 9, 16, 25]

#Criamos a spline cúbica( a nossa curva matemática suave)
spline = CubicSpline(x_pontos, y_pontos)

# Escolhemos pontos intermediários ( no meio dos originais) para testar a curva
x_novos = np.array([1.5, 2.5, 3.5,4.5])

# Calculamos os valores de y para esses novos pontos usando a nossa spline
y_interpolados = spline(x_novos)

# Exibimos os pontos de teste e os resultados encontrados
print("Novos pontos de x onde calculamos:")
print(x_novos)

print("\nResultados interpolados para y:")
y_interpolados
```

Exercício 19 - Realize a transformada de Fourier de um sinal discreto composto de senos e cossenos.
A Transformada de Fourier funciona como um separador de ingredientes para ondas.
Imagine que você tem uma vitamina de banana com morango (que seria o nosso sinal misturado de seno e cosseno). A Transformada de Fourier faz o papel de "desmanchar" essa vitamina, revelando exatamente a quantidade de banana e de morango (as frequências de cada onda) que foram usadas na receita original.
Neste exercício, usei o SciPy para calcular a Transformada Rápida de Fourier (FFT) de um sinal discreto. Primeiro, utilizei o comando np.linspace do NumPy para gerar 8 pontos de tempo igualmente espaçados (que serão os momentos onde mediremos o nosso sinal). Em seguida, criei um sinal composto misturando ondas de seno e cosseno e apliquei a função fft do submódulo scipy.fft. Essa função funciona como um 'separador de ingredientes', desmontando a nossa mistura para revelar na tela as frequências que formavam o sinal original. Para finalizar, usei o np.round para exibir os resultados de forma limpa.






```python
import numpy as np
from scipy.fft import fft

# Criamos um intervalo de tempo curto com 8 pontos discretos
t = np.linspace(0, 1, 8, endpoint=False)

# Geramos o sinalmisturando uma onda de seno e outra de coseno
sinal = np.sin(2 * np.pi * t ) + np.cos(4 * np.pi * t)

# Calcilamos a Transformada de Fourier usando o SciPy
frequencias = fft(sinal)

# Exibimos o sinal e o resultado arrendodado para limpar a tela
print("Sinal original (domínio do tempo):")
print(np.round(sinal, 4))

print("\nTransformada de Fourier (frequências detectadas):")
np.round(frequencias, 4)
```

Exercício 20 -  Resolva um sistema de equações lineares com três variáveis utilizando SciPy

Como a lista não traz um sistema específico, peguei um exemplo didático perfeito que gera resultados inteiros e limpos no seu Google Colab.  1x + 1y + 1z = 6 \\ 2x + 1y - 1z = 1 \\ 1x - 1y + 3z = 8

Para resolver, usamos a função linalg.solve do submódulo scipy.lnlag, que é a ferramenta especializada do SciPy para resolver sistemas da forma de algebra linear.
Criei a matriz A com os pesos das variáveis, o vetor B com as igualdades das fórmulas, e o SciPy encontrou os valores exatos que resolvem o sistema de uma vez só: 1, 2 e 3







```python
import numpy as np
from scipy import linalg

# Definimos a matriz 'A' com os coeficientes  das três equações
A= np.array([[1, 1, 1],
            [2, 1, -1],
            [1, -1, 3]])

# Definimos o vetor 'B' com os resultados das equações
B= np.array([6, 1, 8])

# Resolvemos o sitema linear usando o SciPy
solucao = linalg.solve(A,B)

# Exibimos os resultados encontrados para [x,y,z]
solucao
```

No Pandas, fazemos essa estimativa inteligente de forma automática com a função df.interpolate()
Imagine que você está medindo a altura de uma planta durante a semana:
Na segunda-feira, ela media 10 cm.
Na terça-feira, você esqueceu de medir.
Na quarta-feira, ela media 12 cm.
Se você tivesse que chutar qual era a altura dela na terça-feira, você provavelmente diria 11 cm, certo?
Isso é a interpolação linear!

Neste exercício, criei um DataFrame contendo uma série de alturas ao longo da semana, deixando uma falha (representada por np.nan) na terça-feira. Para resolver isso de forma inteligente, utilizei o método interpolate() do Pandas para realizar uma interpolação linear. Em vez de sobrescrever a coluna original, atribuí os dados interpolados a uma nova coluna chamada Temperatura. Isso nos permitiu comparar o antes e o depois lado a lado: na coluna Altura original, a terça-feira permanece como NaN, enquanto na nova coluna o Pandas calculou de forma precisa o valor intermediário de 11.0.


```python
import numpy as np
import pandas as pd

# Criamos um DataFrame com alguns dados faltando(representados por np.nan)
dados = {
    'Dia': ['Segunda', 'Terça', 'Quarta', 'Quinta', 'Sexta'],
    'Altura': [10, np.nan, 12, 11, 13]
}
df = pd.DataFrame(dados)

# Preenchemos as lacunas de Altura usando interpolação linear e atribuímos à coluna 'Temperatura'
df['Temperatura'] = df['Altura'].interpolate()

# Exibimos o DataFrame com as lacunas preenchidas
df
```

Exercício 1 - Crie um gráfico de linha para a função y = x³ no intervalo de x = -10 até x = 10

Aqui, o Python vai desenhar um gráfico de linha para mostrar visualmente como os valores crescem de forma curva quando elevamos os números ao cubo. Para isso vamos usar a biblioteca Matplotilib (em conjunto com o NumPy) para gerar o gráfico de linha da função cúbica y  = x^3 no intervalo de x = -10 até x = 10 .

Para desenhar a nossa curva, usamos o comando np.linspace(-10, 10, 100) para criar uma sequência de 100 números bem distribuídos começando em -10 e terminando em 10, que serão a base do nosso gráfico (o eixo horizontal X).

Depois, fazemos a conta matemática y = x ** 3, que pega cada um desses números e calcula o seu valor ao cubo para descobrir a altura do gráfico (o eixo vertical Y)
. Por fim, o comando plt.plot(x, y) junta esses pontos criando uma linha contínua e suave, enquanto os comandos plt.title, plt.xlabel e plt.ylabel colocam as etiquetas para que qualquer pessoa entenda o que o gráfico está mostrando, exibindo a imagem final na tela através do comando plt.show()

Exercício 8 - Plote um gráfico de linha com intervalos de confiança (sombramento) para a função y = sin(x).

O que é e por que usamos o Sombramento de Intervalo de Confiança?
Na ciência de dados, as medições ou previsões quase nunca são 100% exatas; elas costumam vir com uma margem de erro ou incerteza.
Para representar isso de forma clara, nós desenhamos a curva principal no centro e criamos uma faixa sombreada ao redor dela. Essa faixa mostra ao leitor o limite inferior e superior onde os dados reais podem variar.
No Matplotlib, fazemos isso usando uma função fantástica chamada plt.fill_between()!
Neste exercício, criei um gráfico de linha para a função seno (np.sin(x)), que é aquela função matemática que gera uma curva parecida com uma onda suave, subindo e descendo. O legal de usar o NumPy (np.) é que ele calcula o seno de todos os nossos 100 pontos de uma vez só.
O destaque do exercício foi desenhar uma 'margem de erro' ao redor dessa onda (o intervalo de confiança). Para fazer esse sombreado, usei a função plt.fill_between() com uma transparência bem suave (alpha=0.25), criando um efeito visual muito bonito e profissional sem poluir o gráfico e o plt.show() exibe na tela.




```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Geramos os dados de x e calculamos o seno (a onda que sobe e desce) de uma vez só
x = np.linspace(0, 10, 100)
y = np.sin(x)

# 2. Criamos a margem de erro (0.2 para cima e para baixo da onda)
y_inferior = y - 0.2
y_superior = y + 0.2

# 3. Desenhamos a linha principal e a sombra suave ao redor dela
plt.plot(x, y, color='green', label='y = sin(x)')
plt.fill_between(x, y_inferior, y_superior, color='blue', alpha=0.25, label='Margem de Erro (±0.2)')

# 4. Título e legenda
plt.title('Onda Senoidal com Intervalo de Confiança Sombreado')
plt.legend()

# 5. Mostramos o gráfico
plt.show()
```

. Ele indica ao Python a partir de qual altura cada barra nova deve começar a ser desenhada, evitando que elas fiquem sobrepostas a partir do chão. O segundo grupo usa o primeiro como base (bottom=grupo1)
, e o terceiro grupo é empilhado usando a soma das duas primeiras barras como apoio (bottom=grupo1 + grupo2). Utilizei arrays do NumPy para podermos somar as listas de dados de forma direta. E o plt.show() para mostrar o resultado na tela.

 O que é a Função $y = 1/x$?Essa é uma função racional (ou de proporção inversa). Ela mostra o que acontece quando dividimos o número 1 por valores cada vez maiores:Se $x = 1$, então $y = 1/1 = 1$ (valor alto).Se $x = 2$, então $y = 1/2 = 0.5$ (caiu pela metade).Se $x = 100$, então $y = 1/100 = 0.01$ (um valor quase colado no zero).Visualmente, isso gera uma curva descendente muito bonita chamada de hipérbole, que despenca rapidamente no começo e depois vai se aproximando suavemente do zero sem nunca tocá-lo!

 Neste exercício, plotamos a função inversa $y = 1/x$. A lógica do código funciona passo a passo assim:import matplotlib.pyplot as plt e import numpy as np: Importamos as duas bibliotecas essenciais para gerar os dados (NumPy) e desenhar o gráfico (Matplotlib).x = np.arange(1, 101): Usamos o comando np.arange do NumPy para criar uma sequência de números inteiros começando em 1 e indo até 100 (colocamos o limite de paragem em 101 porque o Python nunca inclui o último número).y = 1 / x: Aplicamos a fórmula matemática. O Python pega o número 1 e divide por cada um dos valores de X que criamos na linha anterior, gerando as alturas correspondentes para o eixo Y.plt.plot(x, y): Pega todos esses pares de coordenadas (X, Y) e traça a linha contínua do gráfico na tela.plt.title(), plt.xlabel() e plt.ylabel(): Adicionam os textos de identificação (título principal e nomes dos eixos X e Y) para que qualquer pessoa consiga entender o que o gráfico representa. E o plt.show() exibe na tela o resultado.


```python
import matplotlib.pyplot as plt
import numpy as np

# Criamos os valores de x variando de 1 até 100
# O np.arange() gera uma sequência de números
# O último valor não é incluído, por isso usamos 101
x = np.arange(1,101)

# Calculamos os vetores de y de acordo com a função y = 1/x
# Cada valor de x será utilizado como denominador
y = 1/x

# Criamos o gráfico de linha
# O eixo x recebe os valores de x
# O eixo y recebe os valores calculados de y
plt.plot(x,y)

# Adicionamos um título ao gráfico
plt.title('Função y = 1/x')

# Identificamos o eixo x
plt.xlabel('VAlores de X')

# Identificamos o eixo y
plt.ylabel('Valores de Y')

# Exibimos o gráfico na tela
plt.show()
```

Exercício 18 - Crie um gráfico de barras com valores positivos e negativos e use diferentes cores para cada tipo de valor.

Neste exercício, criei um gráfico de barras com valores positivos e negativos
. O grande diferencial foi a utilização de uma compreensão de lista (list comprehension) para definir as cores dinamicamente: as colunas positivas foram coloridas de verde (forestgreen) e as negativas de vermelho (crimson). Também adicionei uma linha de referência horizontal no ponto zero com o comando plt.axhline(0), facilitando a identificação visual de ganhos e perdas ao longo do eixo Y.



```python
import matplotlib.pyplot as plt
import numpy as np

# Definimos 6 categorias e valores (misturando positivos e negativos)
categorias = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun']
valores = [15, -8, 22, -12, 7, -15]

# Criamos uma lista de cores inteligente!
# Ele percorre cada 'v' em valores: se for maior que 0 é verde ('forestgreen'), senão é vermelho ('crimson')
cores = ['forestgreen' if v > 0 else 'crimson' for v in valores]

# Criamos a figura
plt.figure(figsize=(8, 5))

# Desenhamos o gráfico de barras passando a nossa lista de cores personalizada
plt.bar(categorias, valores, color=cores, edgecolor='black', width=0.6)

# Adiciona uma linha horizontal preta exatamente no ponto Y = 0 (ajuda a destacar a divisão)
plt.axhline(0, color='black', linewidth=1, linestyle='-')

# Personalizações básicas do gráfico
plt.title('Gráfico de Barras com Saldos Positivos e Negativos', fontsize=12, fontweight='bold')
plt.xlabel('Meses')
plt.ylabel('Saldo Financeiro')
plt.grid(True, axis='y', linestyle=':', alpha=0.6) # Grade apenas horizontal

# Exibimos o resultado final
plt.show()
```

Exercício 19 -Plote um gráfico de linha com duas séries e adicione uma legenda para diferenciá-las.

Até agora, a maioria dos nossos gráficos de linha mostrava apenas uma única informação por vez. O desafio do Exercício 19 é aprender a comparar duas séries de dados diferentes em uma única tela
.
Para que quem está olhando saiba qual linha representa o quê, vamos usar uma ferramenta indispensável: a legenda
!
A regra do Matplotlib para criar legendas é muito simples e automática:
Dentro de cada comando plt.plot(), nós adicionamos o parâmetro label='Nome da Série'
.
No final, antes de mostrar o gráfico, nós chamamos a função plt.legend()
. O Python vai ler os nomes que você escolheu e montar um quadrinho explicativo perfeito no canto da tela.

Neste exercício, aprendi a plotar múltiplas séries de dados em um mesmo gráfico de linhas
. Utilizei as funções matemáticas de Seno e Cosseno geradas pelo NumPy
. O elemento principal deste desafio foi o uso da função plt.legend(), que coleta automaticamente os textos informados no parâmetro label de cada comando plt.plot() e gera uma caixa explicativa na tela
, tornando a comparação entre as diferentes curvas imediata e legível.

```python
import matplotlib.pyplot as plt
import numpy as np

# Geramos os valores de X variando de 0 a 10 (com 100 pontos para a curva ficar lisa)
x = np.linspace(0, 10, 100)

# Criamos as duas séries de dados diferentes
y1 = np.sin(x)  # Primeira série: Seno
y2 = np.cos(x)  # Segunda série: Cosseno

# Criamos a figura
plt.figure(figsize=(8, 5))

# Desenhamos as duas séries juntas no mesmo gráfico!
# Usamos cores contrastantes e definimos o 'label' de cada uma para a legenda
plt.plot(x, y1, color='dodgerblue', linewidth=2, label='Seno')
plt.plot(x, y2, color='tomato', linewidth=2, linestyle='--', label='Cosseno')

# Personalizações clássicas de títulos e eixos
plt.title('Gráfico de Linhas com Múltiplas Séries', fontsize=12, fontweight='bold')
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')
plt.grid(True, linestyle=':', alpha=0.6)

# O TOQUE FINAL: Chamamos a legenda para identificar cada linha!
plt.legend()

# Exibimos o resultado na tela
plt.show()
```

Exercício 20 - Adicione uma grade com espaçamento personalizado a um gráfico de linha para y = x^2.

Neste exercício final, plotei a função quadrática $y = x^2$2 e aprendi a personalizar o espaçamento da grade de fundo2. Como o Matplotlib posiciona as linhas da grade exatamente sobre os marcadores de escala dos eixos (ticks)56, utilizei os comandos plt.xticks() e plt.yticks() combinados com o np.arange() para forçar marcações de 2 em 2 unidades no eixo X e de 10 em 10 unidades no eixo Y6. Por fim, ativei a grade com plt.grid()34, resultando em um controle milimétrico e personalizado sobre o quadriculado do gráfico.



```python
import matplotlib.pyplot as plt
import numpy as np

# Geramos dados de X variando de -10 a 10 e calculamos y = x^2 (nossa parábola!)
x = np.linspace(-10, 10, 100)
y = x ** 2

# Criamos a figura
plt.figure(figsize=(8, 5))

# Desenhamos a curva da parábola em roxo escuro (purple)
plt.plot(x, y, color='purple', linewidth=2.5, label='y = x²')

# O GRANDE TRUQUE: Definimos o espaçamento personalizado dos eixos (ticks)
# Eixo X: Marcadores de 2 em 2 (começando em -10 e indo até 10)
plt.xticks(np.arange(-10, 11, 2))

# Eixo Y: Marcadores de 10 em 10 (começando em 0 e indo até 100)
plt.yticks(np.arange(0, 101, 10))

# Ativamos a grade (ela seguirá automaticamente o espaçamento personalizado dos ticks!)
# Usamos 'both' para aplicar nos dois eixos, estilo tracejado (--) e cor cinza sutil
plt.grid(True, which='both', linestyle='--', color='gray', alpha=0.5)

# Personalizações finais de títulos e rótulos
plt.title('Gráfico de Linha com Grade de Espaçamento Personalizado', fontsize=12, fontweight='bold')
plt.xlabel('Eixo X (Passo de 2 em 2)')
plt.ylabel('Eixo Y (Passo de 10 em 10)')
plt.legend()

# Exibimos o resultado final
plt.show()
```
